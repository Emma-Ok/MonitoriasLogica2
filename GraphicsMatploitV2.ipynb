# Acercamiento a Matplotlib.

*¿Qué es Matplotlib?* Matplotlib es una librería de Python utilizada para crear gráficos estáticos, animados e interactivos. Es una de las herramientas más poderosas y flexibles en el ecosistema de visualización de datos de Python. Con Matplotlib, puedes crear todo tipo de gráficos, desde simples gráficos de líneas hasta complejos gráficos en 3D.

*¿Por qué es útil?* Matplotlib es especialmente útil en el análisis de datos, ya que permite visualizar grandes volúmenes de información de manera comprensible. Ya sea en ciencia de datos, machine learning o análisis financiero, la capacidad de crear gráficos claros y personalizables ayuda a comunicar resultados de forma efectiva.



# Estructura básica de Matplotlib

*Componentes Principales*
  * pyplot: Es el módulo más utilizado en Matplotlib. Proporciona una interfaz simple y fácil de usar para crear gráficos rápidamente
  * figure: Es el contenedor principal de un gráfico. Representa la ventana o el espacio donde se dibuja el gráfico.
  * axes: Son los ejes del gráfico y contienen los datos visualizados. Cada figura puede tener múltiples ejes.

```python
from typing import List

class AlgoritmosOrdenamiento:
    @staticmethod
    def bubble_sort(arr: List[int]) -> List[int]:
        n = len(arr)
        arr_copy = arr.copy()
        for i in range(n):
            for j in range(0, n - i - 1):
                if arr_copy[j] > arr_copy[j + 1]:
                    arr_copy[j], arr_copy[j + 1] = arr_copy[j + 1], arr_copy[j]
        return arr_copy

    @staticmethod
    def selection_sort(arr: List[int]) -> List[int]:
        arr_copy = arr.copy()
        for i in range(len(arr_copy)):
            min_idx = i
            for j in range(i + 1, len(arr_copy)):
                if arr_copy[j] < arr_copy[min_idx]:
                    min_idx = j
            arr_copy[i], arr_copy[min_idx] = arr_copy[min_idx], arr_copy[i]
        return arr_copy

    @staticmethod
    def insertion_sort(arr: List[int]) -> List[int]:
        arr_copy = arr.copy()
        for i in range(1, len(arr_copy)):
            key = arr_copy[i]
            j = i - 1
            while j >= 0 and arr_copy[j] > key:
                arr_copy[j + 1] = arr_copy[j]
                j -= 1
            arr_copy[j + 1] = key
        return arr_copy

    @staticmethod
    def merge_sort(arr: List[int]) -> List[int]:
        if len(arr) <= 1:
            return arr

        mid = len(arr) // 2
        left = AlgoritmosOrdenamiento.merge_sort(arr[:mid])
        right = AlgoritmosOrdenamiento.merge_sort(arr[mid:])

        return AlgoritmosOrdenamiento._merge(left, right)

    @staticmethod
    def _merge(left: List[int], right: List[int]) -> List[int]:
        result = []
        i = j = 0

        while i < len(left) and j < len(right):
            if left[i] <= right[j]:
                result.append(left[i])
                i += 1
            else:
                result.append(right[j])
                j += 1

        result.extend(left[i:])
        result.extend(right[j:])
        return result

    @staticmethod
    def quick_sort(arr: List[int]) -> List[int]:
        if len(arr) <= 1:
            return arr
        else:
            pivot = arr[0]
            less = [x for x in arr[1:] if x <= pivot]
            greater = [x for x in arr[1:] if x > pivot]
            return AlgoritmosOrdenamiento.quick_sort(less) + [pivot] + AlgoritmosOrdenamiento.quick_sort(greater)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import time

def medir_tiempo_ordenamiento(algoritmo, arr):
    inicio = time.time()
    algoritmo(arr)
    fin = time.time()
    return fin - inicio

# Configuración de la visualización
plt.style.use('seaborn-v0_8-notebook')
fig = plt.figure(figsize=(20, 12))

# Lista de algoritmos a evaluar
algoritmos = {
    'Bubble Sort': AlgoritmosOrdenamiento.bubble_sort,
    'Selection Sort': AlgoritmosOrdenamiento.selection_sort,
    'Insertion Sort': AlgoritmosOrdenamiento.insertion_sort,
    'Merge Sort': AlgoritmosOrdenamiento.merge_sort,
    'Quick Sort': AlgoritmosOrdenamiento.quick_sort
}

# Definir tres rangos diferentes de tamaños para mejor visualización
small_sizes = np.linspace(100, 1000, 10, dtype=int)
medium_sizes = np.linspace(1000, 2000, 10, dtype=int)
large_sizes = np.linspace(2000, 5000, 5, dtype=int)

# Tamaños de arrays a probar
all_sizes = np.concatenate([small_sizes, medium_sizes, large_sizes])
tiempos = {nombre: [] for nombre in algoritmos}

# Medir tiempos para cada algoritmo y tamaño
for size in all_sizes:
    arr = np.random.randint(0, 10, size=size)
    for nombre, algoritmo in algoritmos.items():
        tiempo = medir_tiempo_ordenamiento(algoritmo, arr.tolist())
        tiempos[nombre].append(tiempo)

fig1 = plt.figure(figsize=(20, 10))
fig1.suptitle('Análisis Individual de Algoritmos de Ordenamiento', fontsize=16)
# Crear subplots individuales
for idx, (nombre, algoritmo) in enumerate(algoritmos.items(), 1):
    plt.subplot(2, 3, idx)
    plt.plot(all_sizes, tiempos[nombre], '-o')
    plt.title(f'{nombre}', fontsize=12)
    plt.xlabel('Tamaño del array', fontsize=10)
    plt.ylabel('Tiempo (segundos)', fontsize=10)
    plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Crear figura para gráfico comparativo
fig2 = plt.figure(figsize=(15, 8))
for nombre, algo_info in algoritmos.items():
    plt.plot(all_sizes, tiempos[nombre], '-o', label=nombre)
plt.title('Comparación de Algoritmos de Ordenamiento', fontsize=14)
plt.xlabel('Tamaño del array', fontsize=12)
plt.ylabel('Tiempo (segundos)', fontsize=12)
plt.legend(fontsize=10, loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# Análisis de resultados
print("Análisis de rendimiento:")
for size in [1000, 2000, 5000]:
    print(f"\nPara arrays de tamaño {size}:")
    idx = np.where(all_sizes == size)[0][0]
    sorted_algorithms = sorted(algoritmos.keys(), key=lambda x: tiempos[x][idx])
    for i, algo in enumerate(sorted_algorithms, 1):
        print(f"{i}. {algo}: {tiempos[algo][idx]:.6f} segundos")
```

## Barras vs. Barras apiladas

- Útil para comparar categorías. Las apiladas muestran composición total.

Tip: usa `plt.xticks(rotation=...)` si las etiquetas son largas.

Dato curioso: Elegir colores accesibles (daltónicos) mejora la inclusión; `tab10` es una paleta segura.

```python
import numpy as np
import matplotlib.pyplot as plt

categorias = ['A', 'B', 'C', 'D']
val1 = [5, 7, 3, 4]
val2 = [2, 3, 4, 6]
idx = np.arange(len(categorias))

fig, ax = plt.subplots(1, 2, figsize=(10,4), sharey=True)

# Barras lado a lado
ax[0].bar(idx - 0.2, val1, width=0.4, label='Serie 1')
ax[0].bar(idx + 0.2, val2, width=0.4, label='Serie 2')
ax[0].set_xticks(idx)
ax[0].set_xticklabels(categorias)
ax[0].set_title('Barras lado a lado')
ax[0].legend()
ax[0].grid(axis='y', alpha=0.3)

# Barras apiladas
ax[1].bar(idx, val1, label='Serie 1')
ax[1].bar(idx, val2, bottom=val1, label='Serie 2')
ax[1].set_xticks(idx)
ax[1].set_xticklabels(categorias)
ax[1].set_title('Barras apiladas')
ax[1].legend()
ax[1].grid(axis='y', alpha=0.3)

fig.suptitle('Comparación de barras')
fig.tight_layout()
plt.show()
```

### Histogramas y boxplot: distribución y outliers

- Histograma: forma de la distribución.
- Boxplot: mediana, cuartiles y outliers.

Tip: Ajusta `bins` según la cantidad de datos; prueba la regla de Freedman–Diaconis.

Dato curioso: Los boxplots fueron popularizados por John Tukey, pionero del análisis exploratorio de datos.

```python
data = np.concatenate([
    np.random.normal(0, 1, 500),
    np.random.normal(5, 1.5, 300),
    np.array([12, 13, 15])  # outliers
])

fig, ax = plt.subplots(1, 2, figsize=(10,4))
ax[0].hist(data, bins=30, color='tab:green', alpha=0.7)
ax[0].set_title('Histograma')
ax[0].set_xlabel('Valor')
ax[0].set_ylabel('Frecuencia')
ax[0].grid(alpha=0.3)

ax[1].boxplot(data, vert=True, patch_artist=True,
              boxprops=dict(facecolor='lightblue', color='navy'),
              medianprops=dict(color='red'))
ax[1].set_title('Boxplot')
ax[1].set_ylabel('Valor')
ax[1].grid(axis='y', alpha=0.3)

fig.suptitle('Distribución y outliers')
fig.tight_layout()
plt.show()
```

### Anotaciones, flechas y resaltados

- Ideal para comunicar un insight concreto en el gráfico.

Tip: Usa `bbox` y `arrowprops` para destacar información.

Dato curioso: Una buena anotación puede ahorrar varias frases de texto.

```python
x = np.linspace(0, 10, 200)
y = np.sin(x) + 0.3*np.random.randn(200)

plt.figure(figsize=(7,4))
plt.plot(x, y, label='Señal')

# Buscar un pico simple
peak_idx = np.argmax(y)
peak_x, peak_y = x[peak_idx], y[peak_idx]

plt.scatter([peak_x], [peak_y], color='red', zorder=5)
plt.annotate('Pico máximo', xy=(peak_x, peak_y), xytext=(peak_x+0.5, peak_y+0.5),
             arrowprops=dict(arrowstyle='->', color='red'),
             bbox=dict(boxstyle='round,pad=0.3', fc='yellow', alpha=0.3))

# Resaltado de región
plt.axvspan(3, 5, color='orange', alpha=0.2, label='Región de interés')
plt.title('Anotaciones y resaltados')
plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

###  Ejes secundarios (twinx) y escalas logarítmicas

- `twinx()` permite dos escalas distintas sobre el mismo eje x.
- Escala log (`set_yscale('log')`) es útil para datos con órdenes de magnitud.

Dato curioso: Muchas leyes naturales siguen distribuciones con colas largas donde la escala log es clave.

```python
x = np.linspace(1, 100, 100)
y1 = np.sqrt(x)
y2 = x**2

fig, ax1 = plt.subplots(figsize=(7,4))
line1 = ax1.plot(x, y1, 'g-', label='sqrt(x)')
ax1.set_xlabel('x')
ax1.set_ylabel('sqrt(x)', color='g')
ax1.tick_params(axis='y', labelcolor='g')

ax2 = ax1.twinx()
line2 = ax2.plot(x, y2, 'b--', label='x^2')
ax2.set_ylabel('x^2', color='b')
ax2.tick_params(axis='y', labelcolor='b')
ax2.set_yscale('log')  # escala log para x^2

# Construir leyenda conjunta
lines = line1 + line2
labels = [l.get_label() for l in lines]
ax1.legend(lines, labels, loc='upper left')

plt.title('Eje secundario y escala log')
fig.tight_layout()
plt.show()
```
