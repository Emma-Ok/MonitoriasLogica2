## 1. ¿Qué es la Programación Dinámica?

La programación dinámica es un método algorítmico desarrollado por **Richard Bellman en 1953** para resolver problemas de optimización complejos. Se trata de una técnica que **reduce el tiempo de ejecución de un algoritmo mediante la utilización de subproblemas superpuestos y subestructuras óptimas**.

### Definición Formal

La programación dinámica es un método para reducir el tiempo de ejecución de un algoritmo mediante la utilización de subproblemas superpuestos y subestructuras óptimas. El matemático Richard Bellman inventó la programación dinámica en 1953 que se utiliza para optimizar problemas complejos que pueden ser discretizados y secuencializados.
<br>
En otras palabras la programación dinámica es una técnica de optimización que resuelve problemas complejos dividiéndolos en subproblemas más simples, almacenando los resultados para evitar recálculos y se basa en dos principios fundamentales
Subestructura óptima: La solución óptima contiene soluciones óptimas de subproblema
<br/>
Subproblemas superpuestos: Los mismos subproblemas se resuelven múltiples veces

### Características Fundamentales

**Principio de Optimalidad de Bellman**: Dado el estado actual, una política óptima para las etapas restantes es independiente de la política adoptada en etapas anteriores, por ende, la decisión inmediata óptima solo depende del estado actual y no de cómo se llegó ahí

**Dos Propiedades Esenciales:**

1. **Subestructura Óptima**: Se pueden usar soluciones óptimas de subproblemas para encontrar la solución óptima del problema en su conjunto<br>
 *La solución óptima contiene soluciones óptimas de subproblema*
<br>


2. **Subproblemas Superpuestos**: Decir que un problema tiene subproblemas superpuestos es decir que se usa un mismo subproblema para resolver diferentes problemas mayores <br>
 *Los mismos subproblemas se resuelven múltiples veces*

### Enfoques de Implementación

**Top-down (Memoización)**: El problema se divide en subproblemas, y estos se resuelven recordando las soluciones por si fueran necesarias nuevamente. Es una combinación de memoización y recursión

**Bottom-up (Constructivo)**: Todos los problemas que puedan ser necesarios se resuelven de antemano y después se usan para resolver las soluciones a problemas mayores

## 5. ¿Cómo se codifican soluciones con programación dinámica?

La resolución de un problema mediante programación dinámica implica los siguientes pasos: Definir los subproblemas: Un problema grande se divide en pequeños subproblemas. Resolver los subproblemas: Esto implica resolver el subproblema identificado, lo que puede hacerse utilizando la recursividad o la iteración. Construir la solución al problema original: La solución al problema grande se construye a partir de los subproblemas que ya se han calculado

## 6. Caso de Ejemplo: Subsecuencia Común Más Larga (LCS)

### Descripción del Problema

El Longest Common Subsequence (LCS) es un problema clásico en informática. El objetivo es encontrar la subsecuencia común más larga ("Longest Common Subsequence") de dos secuencias

**Definición**: El problema de subsecuencia común más larga, se trata de encontrar una subsecuencia más larga que es común en un conjunto de secuencias. Es diferente del problema de substring común más largo; a diferencia de los substrings, las subsecuencias no necesitan tener posiciones consecutivas en la secuencia original

### Aplicaciones Prácticas

El problema de la Longest Common Subsequence es importante en todos los ámbitos en los que se trabaje con secuencias derivadas unas de otras. Las soluciones para encontrar la LCS se utilizan, por ejemplo, para examinar textos en busca de similitudes y diferencias. De esta forma puede detectarse el plagio. La conocida utilidad "diff", que muestra los cambios en archivos de código fuente, también se basa en el problema LCS

### Ejemplo Práctico

**Cadenas de entrada:**
- X = "AGGTAB"
- Y = "GXTXAYB"

**Proceso:**
1. Comparamos caracteres de ambas cadenas
2. Si coinciden: LCS[i][j] = LCS[i-1][j-1] + 1
3. Si no coinciden: LCS[i][j] = max(LCS[i-1][j], LCS[i][j-1])

**Resultado:** La LCS es "GTAB" con longitud 4

### Características del Problema LCS

El problema de LCS tiene una estructura óptima: el problema puede ser dividido en subproblemas más pequeños y simples, el cual puede ser dividido en subproblemas más simples y así hasta que la solución llega a ser trivial

El problema de LCS también tiene subproblemas que se traslapan: las soluciones de subproblemas de nivel mayor utilizan soluciones de subproblemas menores


### Análisis de Complejidad

**Enfoque Recursivo Puro:**
- Tiempo: O(2^min(m,n)) - Exponencial
- Espacio: O(min(m,n)) - Profundidad de recursión

**Enfoque con Programación Dinámica:**
- Tiempo: O(m × n) - Polinomial
- Espacio: O(m × n) - Para la tabla DP

Para el caso de dos secuencias de n y m elementos, el tiempo de ejecución para la programación dinámica es de O(n × m)


### Análisis Teórico

| Enfoque | Complejidad Temporal | Complejidad Espacial | Características |
|---------|---------------------|---------------------|-----------------|
| Recursivo Puro | O(2^min(m,n)) | O(min(m,n)) | Muy ineficiente, solo para casos pequeños |
| Memoización (Top-down) | O(m×n) | O(m×n) | Eficiente, fácil de implementar |
| Bottom-up | O(m×n) | O(m×n) | Eficiente, no usa recursión |
| Optimizado en Espacio | O(m×n) | O(min(m,n)) | Eficiente en memoria |

### Ventajas de Programación Dinámica sobre Recursión Pura

1. **Eliminación de Cálculos Redundantes**: Esto se puede evitar guardando las soluciones que ya hemos calculado. Entonces, si necesitamos resolver el mismo problema más tarde, podemos obtener la solución de la lista de soluciones calculadas y reutilizarla

2. **Reducción Dramática de Complejidad**: La función resultante tiene complejidad O(n), en lugar de 2 a la n

3. **Escalabilidad**: Permite resolver problemas de mayor tamaño que serían imposibles con recursión pura

```python

import time
import numpy as np
import pandas as pd
import random
import matplotlib.pyplot as plt
import seaborn as sns
from functools import lru_cache
from scipy.stats import gmean
from scipy import stats
import warnings

# Configuración visual
warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)
```

```python
# BLOQUE 2: Clase optimizada para comparación de algoritmos Fibonacci
class FibonacciComparator:
    def __init__(self):
        self.results_df = None

    def fibonacci_recursive(self, n):
        if n <= 1:
            return n
        return self.fibonacci_recursive(n - 1) + self.fibonacci_recursive(n - 2)

    @lru_cache(maxsize=None)
    def fibonacci_memoization(self, n):
        if n <= 1:
            return n
        return self.fibonacci_memoization(n - 1) + self.fibonacci_memoization(n - 2)

    def fibonacci_iterative(self, n):
        if n <= 1:
            return n
        a, b = 0, 1
        for _ in range(2, n + 1):
            a, b = b, a + b
        return b

    def measure_time(self, func, n, max_time=3.0):
        try:
            start = time.perf_counter()
            result = func(n)
            end = time.perf_counter()
            elapsed = end - start
            return elapsed if elapsed <= max_time else None
        except Exception:
            return None

    def _get_complexity(self, algorithm_name):
        return {
            'Recursivo': 'O(2^n)',
            'Memoización': 'O(n)',
            'Iterativo': 'O(n)'
        }[algorithm_name]

    def collect_data(self, n_min=5, n_max=100, max_time=3.0):
        algorithms = {
            'Recursivo': self.fibonacci_recursive,
            'Memoización': self.fibonacci_memoization,
            'Iterativo': self.fibonacci_iterative
        }
        data = []
        for n in range(n_min, n_max + 1):
            for name, func in algorithms.items():
                if hasattr(func, 'cache_clear'):
                    func.cache_clear()
                exec_time = self.measure_time(func, n, max_time)
                if exec_time is not None:
                    data.append({
                        'n': n, 'Algoritmo': name,
                        'Tiempo': exec_time,
                        'Complejidad': self._get_complexity(name)
                    })
        self.results_df = pd.DataFrame(data)
        return self.results_df


    def plot_comparison(self):
        if self.results_df is None: return
        g = sns.relplot(
            data=self.results_df, x='n', y='Tiempo',
            hue='Algoritmo', style='Complejidad',
            kind='line', markers=True, linewidth=2.5, height=7, aspect=1.3
        )
        g.ax.set_yscale('log')
        g.ax.set_title('Comparación de Implementaciones de Fibonacci', fontsize=16)
        g.ax.set_xlabel('n')
        g.ax.set_ylabel('Tiempo (s)')
        plt.tight_layout()
        plt.show()

    def print_summary(self):
        if self.results_df is None: return
        print("RESUMEN DE TIEMPOS (Media geométrica en segundos):\n")
        grouped = self.results_df.groupby('Algoritmo')['Tiempo'].agg([gmean, min, max])
        print(grouped.round(6))

        # Speedups comparativos para ciertos valores de n
        for n in [20, 25, 30]:
            n_data = self.results_df[self.results_df['n'] == n]
            if len(n_data) == 3:
                t = n_data.set_index("Algoritmo")["Tiempo"]
                print(f"\nn={n}:")
                print(f"  Memoización: {t['Recursivo']/t['Memoización']:.0f}x más rápido")
                print(f"  Iterativo:   {t['Recursivo']/t['Iterativo']:.0f}x más rápido")

```

```python
def run_fibonacci_analysis(n_min=5, n_max=35):
    comp = FibonacciComparator()
    print(f"Recolectando datos de n={n_min} a n={n_max}...\n")
    results = comp.collect_data(n_min, n_max, max_time=3.0)
    comp.plot_comparison()
    comp.print_summary()
    return comp, results

comp, results = run_fibonacci_analysis(n_min=5, n_max=42)
```

**Programacion dinámica**
El problema consiste en grupo de jugadores que toman
turnos para remover monedas desde una l´ınea de monedas hasta maximizar el valor total de las
monedas recogidas.

Considerando la descripción del problema, para el diseño con aproximación descendente o
memoization, es necesario considerar cuatro condiciones para el diseño del algoritmo:
• Descomponer el problema en subproblemas. Para el contexto del juego Monedas en una línea,
el subproblema podría ser encontrar el valor máximo de una sub-secuencia de monedas.
• Para facilitar el almacenamiento de las soluciones de los subproblemas, es necesario establecer
una matriz T, tal que T[h, k] represente el máximo valor que puede ser obtenido desde la
sub-secuencia de monedas desde la posición h hasta la posición k.
• Al ejecutar el algoritmo, se debe rellenar la matriz desde los subproblemas más pequeños
hasta los subproblemas más grandes, haciendo uso de valores calculados previamente.

```python
def monedas_en_una_linea(monedas):
    n = len(monedas)
    # Crear una tabla DP de tamaño n x n
    dp = [[0] * n for _ in range(n)]

    # Llenar la tabla para subproblemas de tamaño 1 (diagonal principal)
    for i in range(n):
        dp[i][i] = monedas[i]

    # Llenar la tabla para subproblemas más grandes (visión descendente o ascendente)
    for longitud in range(2, n + 1):  # longitud de la subcadena
        for i in range(n - longitud + 1):
            j = i + longitud - 1  # Fin del rango actual
            # Elegir el extremo izquierdo o derecho
            opcion_izquierda = monedas[i] + min(
                dp[i + 2][j] if i + 2 <= j else 0,
                dp[i + 1][j - 1] if i + 1 <= j - 1 else 0
            )
            opcion_derecha = monedas[j] + min(
                dp[i][j - 2] if i <= j - 2 else 0,
                dp[i + 1][j - 1] if i + 1 <= j - 1 else 0
            )
            # Guardar el valor máximo en la tabla
            dp[i][j] = max(opcion_izquierda, opcion_derecha)

    # El resultado final está en dp[0][n-1]
    return dp[0][n - 1], dp
```

```python

sizes = list(range(10, 1000, 20))
times = []

for size in sizes:
    monedas = np.random.randint(1, 100, size).tolist()
    start = time.perf_counter()
    monedas_en_una_linea(monedas)
    end = time.perf_counter()
    times.append(end - start)

# Crear DataFrame con los resultados
df_tiempos = pd.DataFrame({
    "Tamaño": sizes,
    "Tiempo": times
})

# Calcular estadísticas
stats = {
    "media geométrica": gmean(times),
    "mínimo": np.min(times),
    "máximo": np.max(times),
    "desviación estándar": np.std(times)
}

print("Tiempos de ejecución (primeras 5 muestras):")
display(df_tiempos.head())

# Reformatear estadísticas en DataFrame legible
df_stats = pd.DataFrame.from_dict(stats, orient='index', columns=["Tiempo (segundos)"])
df_stats.index = [
    "Media geométrica",
    "Mínimo",
    "Máximo",
    "Desviación estándar"
]

# Redondear los valores a 6 decimales para presentación clara
df_stats = df_stats.round(6)

print("Estadísticas de rendimiento del algoritmo:")
display(df_stats)
```

Maxima suma de los nodos con Backtracking

```python
def inicio(n): # n es el numero de nodos
  nodo = {}
  for i in range(n + 1):
    nodo[i] = []
  return nodo

def estructura_arbol_no_dirigido(nodo,pos):
  for k in range(0,len(pos)):
    nodo[pos[k][0]].append(pos[k][1])
    nodo[pos[k][1]].append(pos[k][0])
  return nodo

PD = [0]*50

def dfs_no_dirigido(valores_nodos,nodo,pos_inicial,raiz):
  PD[pos_inicial] = valores_nodos[pos_inicial - 1]
  valor_maximo = 0
  for nodo_hijo in nodo[pos_inicial]:
    if nodo_hijo == raiz:
      continue
    dfs_no_dirigido(valores_nodos,nodo,nodo_hijo,pos_inicial)
    valor_maximo = max(valor_maximo, PD[nodo_hijo])
  PD[pos_inicial] += valor_maximo

def Calcular_Maximos(valores_nodos, nodo):
    dfs_no_dirigido(valores_nodos, nodo, 1, 0)
    return PD[1]
```

Máxima profundidad de un árbol usando Memoization

```python
class Trayectoria_Arbol:
    def __init__(self):
        self.Aristas, self.PD, self.nodo_visitado, self.MaxTray= [], [], [], 0

    def dfs_trayectoria(self, nodo_actual, puntos):
        nodo_hijos, self.registros = [], []
        self.nodo_visitado[nodo_actual] = 1
        for i in self.Aristas[nodo_actual]:
            if not self.nodo_visitado[i[0]]:
                self.dfs_trayectoria(i[0], puntos)
            nodo_hijos.append(self.PD[i[0]] - i[1])
        nodo_hijos.sort(reverse=True)
        maximo_x, maximo_y = 0, 0
        if len(nodo_hijos) >= 2:
            maximo_x = max(maximo_x, nodo_hijos[0])
            maximo_y = max(maximo_y, nodo_hijos[1])
        elif len(nodo_hijos) >= 1:
            maximo_x = max(maximo_x, nodo_hijos[0])
        self.MaxTray = max(self.MaxTray,
                           maximo_x + maximo_y + puntos[nodo_actual])
        self.PD[nodo_actual] = maximo_x + puntos[nodo_actual]

    def MaximaTrayectoria(self, n, puntos, Aristas_Ponderadas):
        self.Aristas = [[] for _ in range(n + 1)]
        self.PD, self.nodo_visitado= [0]*(n + 1), [0]*(n + 1)
        for i in range(n - 1):
            self.Aristas[Aristas_Ponderadas[i][0]].append((
                Aristas_Ponderadas[i][1], Aristas_Ponderadas[i][2]))
            self.Aristas[Aristas_Ponderadas[i][1]].append((
                Aristas_Ponderadas[i][0], Aristas_Ponderadas[i][2]))
        self.dfs_trayectoria(1, puntos)
        return self.MaxTray, self.PD
```

```python
def insertar_adyacencia(relaciones, n):
    nodos_adyacentes = [[] for i in range(n + 1)]
    for i in range(len(relaciones)):
        nodos_adyacentes[relaciones[i][0]].append(relaciones[i][1])
    return nodos_adyacentes

def dfs_pd(nodo, nodos_adyacentes, PD, nodo_visitado):
    if nodo_visitado[nodo]:
        return PD[nodo]

    nodo_visitado[nodo] = True
    max_camino = 0

    for vecino in nodos_adyacentes[nodo]:
        max_camino = max(max_camino, 1 + dfs_pd(vecino, nodos_adyacentes, PD, nodo_visitado))

    PD[nodo] = max_camino
    return PD[nodo]

def Trayectoria_Maxima(nodos_adyacentes, n):
    PD = [0]*(n + 1)
    Valor = 0
    nodo_visitado = [False]*(n + 1)
    for i in range(1, n + 1):
        if not nodo_visitado[i]:
            dfs_pd(i, nodos_adyacentes, PD, nodo_visitado)
    for i in range(1, n + 1):
        Valor = max(Valor, PD[i])
    return Valor, PD

n = 6
relaciones = [[1,4], [1,5], [1,6], [2,3], [2,4], [3,5], [4,5], [4,6], [5,6]]
nodos_adyacentes = insertar_adyacencia(relaciones, n)
a,b = Trayectoria_Maxima(nodos_adyacentes, n)
print(a)
print(b)
```

# Búsqueda de la Trayectoria mínima con GDA ponderada

```python
from collections import defaultdict

class GDA:
    def __init__(self, nodos):
        self.V = nodos
        self.G = defaultdict(list)

    def aristas(self, i, j, w):
        self.G[i].append((j, w))

    def orden_topografico_subgrafo(self, nodo, nodo_predecesor, nodo_sucesor):
        nodo_predecesor[nodo] = True
        if i in self.G:
            for j,_ in self.G[i]:
                if not nodo_predecesor[j]:
                    self.orden_topografico_subgrafo(j, nodo_predecesor, nodo_sucesor)
        nodo_sucesor.append(nodo)

    def orden_topografico_principal(self):
        nodo_predecesor = {v: False for v in range(self.V)}
        nodo_sucesor = []
        for i in range(self.V):
            if not nodo_predecesor[i]:
                self.orden_topografico_subgrafo(i, nodo_predecesor, nodo_sucesor)
        return nodo_sucesor[::-1]

    def Minima_Trayectoria(self, nodo_inicio):
        nodos_sucesores = self.orden_topografico_principal()
        L = {v: float('inf') for v in range(self.V)}
        L[nodo_inicio] = 0
        while nodos_sucesores:
            i = nodos_sucesores.pop()
            if L[i] != float('inf'):
                for j, w in self.G[i]:
                    if L[i] + w < L[j]:
                        L[j] = L[i] + w
        return L



```

```python
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 10)
plt.rcParams['font.size'] = 11

class LCSComparison:
    """
    Análisis optimizado del problema LCS enfocado en visualizaciones clave
    y demostración del proceso algorítmico
    """

    def __init__(self):
        self.algorithms = {
            'Recursivo Puro': self._lcs_recursive,
            'Memoización': self._lcs_memoization,
            'Prog. Dinámica': self._lcs_dp
        }

    def _lcs_recursive(self, X, Y, m=None, n=None):
        """Implementación recursiva pura O(2^min(m,n))"""
        if m is None: m = len(X)
        if n is None: n = len(Y)

        if m == 0 or n == 0:
            return 0

        if X[m-1] == Y[n-1]:
            return 1 + self._lcs_recursive(X, Y, m-1, n-1)
        else:
            return max(
                self._lcs_recursive(X, Y, m-1, n),
                self._lcs_recursive(X, Y, m, n-1)
            )

    def _lcs_memoization(self, X, Y):
        """Implementación con memoización O(m*n)"""
        memo = {}

        def helper(i, j):
            if i == 0 or j == 0:
                return 0
            if (i, j) in memo:
                return memo[(i, j)]

            if X[i-1] == Y[j-1]:
                result = 1 + helper(i-1, j-1)
            else:
                result = max(helper(i-1, j), helper(i, j-1))

            memo[(i, j)] = result
            return result

        return helper(len(X), len(Y))

    def _lcs_dp(self, X, Y):
        """Implementación con programación dinámica O(m*n)"""
        m, n = len(X), len(Y)
        dp = [[0] * (n + 1) for _ in range(m + 1)]

        for i in range(1, m + 1):
            for j in range(1, n + 1):
                if X[i-1] == Y[j-1]:
                    dp[i][j] = dp[i-1][j-1] + 1
                else:
                    dp[i][j] = max(dp[i-1][j], dp[i][j-1])

        return dp[m][n]

    def generate_test_data(self, max_length=14):
        """Genera datos de prueba para el análisis de rendimiento"""
        import random
        random.seed(42)  # Para reproducibilidad

        results = []
        alphabet = 'ABCDEFGHIJKLMNOP'

        print("🔄 Ejecutando análisis de rendimiento...")
        print("-" * 50)

        for length in range(5, max_length + 1):
            # Generar cadenas de prueba
            X = ''.join(random.choices(alphabet[:8], k=length))
            Y = ''.join(random.choices(alphabet[:8], k=length))

            print(f"📏 Longitud {length}: X='{X[:10]}...', Y='{Y[:10]}...'")

            for alg_name, algorithm in self.algorithms.items():
                # Limitar recursivo puro a longitudes pequeñas
                if length > 11 and alg_name == 'Recursivo Puro':
                    continue

                try:
                    start_time = time.time()
                    result = algorithm(X, Y)
                    execution_time = time.time() - start_time

                    results.append({
                        'Longitud': length,
                        'Algoritmo': alg_name,
                        'Tiempo': execution_time,
                        'Resultado': result
                    })

                    print(f"  ✓ {alg_name}: {execution_time:.6f}s (LCS={result})")

                except Exception as e:
                    print(f"  ❌ {alg_name}: Error o timeout")

        return pd.DataFrame(results)

    def demonstrate_algorithm_steps(self, X="AGGTAB", Y="GXTXAYB"):
        """
        Demuestra paso a paso el algoritmo de programación dinámica
        """
        print(f"📊 DEMOSTRACIÓN PASO A PASO: LCS('{X}', '{Y}')")
        print(f"{'='*70}")

        m, n = len(X), len(Y)
        dp = [[0] * (n + 1) for _ in range(m + 1)]

        # Mostrar inicialización
        print("\n1️⃣ INICIALIZACIÓN DE LA TABLA DP:")
        self._print_beautiful_table(dp, X, Y)

        print(f"\n2️⃣ PROCESO DE LLENADO (BOTTOM-UP):")
        print("-" * 50)

        step = 0
        for i in range(1, m + 1):
            for j in range(1, n + 1):
                step += 1
                char_x, char_y = X[i-1], Y[j-1]

                if char_x == char_y:
                    dp[i][j] = dp[i-1][j-1] + 1
                    print(f"Paso {step:2d}: '{char_x}' = '{char_y}' → dp[{i},{j}] = dp[{i-1},{j-1}] + 1 = {dp[i][j]} ✅")
                else:
                    dp[i][j] = max(dp[i-1][j], dp[i][j-1])
                    print(f"Paso {step:2d}: '{char_x}' ≠ '{char_y}' → dp[{i},{j}] = max({dp[i-1][j]}, {dp[i][j-1]}) = {dp[i][j]} ⚠️")

        print(f"\n3️⃣ TABLA DP COMPLETA:")
        self._print_beautiful_table(dp, X, Y)

        # Reconstruir LCS
        lcs_sequence = self._reconstruct_lcs(X, Y, dp)
        print(f"\n4️⃣ RECONSTRUCCIÓN DE LA LCS:")
        print(f"🎯 Subsecuencia común más larga: '{lcs_sequence}'")
        print(f"📏 Longitud: {len(lcs_sequence)}")

        return dp, lcs_sequence

    def _print_beautiful_table(self, dp, X, Y):
        """Imprime una tabla DP visualmente atractiva"""
        m, n = len(X), len(Y)

        # Encabezado con caracteres de Y
        print("\n    ", end="")
        for j in range(n + 1):
            if j == 0:
                print("  ε", end="")
            else:
                print(f"  {Y[j-1]}", end="")
        print()
        print("    " + "―" * (3 * (n + 1)))

        # Filas con caracteres de X
        for i in range(m + 1):
            if i == 0:
                print("ε │ ", end="")
            else:
                print(f"{X[i-1]} │ ", end="")

            for j in range(n + 1):
                print(f"{dp[i][j]:2}", end=" ")
            print()

    def _reconstruct_lcs(self, X, Y, dp):
        """Reconstruye la LCS desde la tabla DP"""
        m, n = len(X), len(Y)
        lcs = []

        i, j = m, n
        while i > 0 and j > 0:
            if X[i-1] == Y[j-1]:
                lcs.append(X[i-1])
                i -= 1
                j -= 1
            elif dp[i-1][j] > dp[i][j-1]:
                i -= 1
            else:
                j -= 1

        return ''.join(reversed(lcs))

    def create_performance_visualization(self, df):
        """
        Crea visualización optimizada de rendimiento
        """
        fig, (ax1) = plt.subplots(figsize=(16, 7))

        # Paleta de colores moderna
        colors = ['#e74c3c', '#3498db', '#2ecc71']

        # Gráfico 1: Comparación de tiempos (escala logarítmica)
        sns.lineplot(data=df, x='Longitud', y='Tiempo', hue='Algoritmo',
                     marker='o', linewidth=3, markersize=8, ax=ax1, palette=colors)

        ax1.set_yscale('log')
        ax1.set_title('⏱️ Comparación de Tiempos de Ejecución', fontsize=14, fontweight='bold', pad=20)
        ax1.set_xlabel('Longitud de las Cadenas', fontsize=12)
        ax1.set_ylabel('Tiempo de Ejecución (segundos, escala log)', fontsize=12)
        ax1.grid(True, alpha=0.3)
        ax1.legend(title='Algoritmo', title_fontsize=11, fontsize=10, loc='upper left')

        # Agregar anotaciones
        if not df.empty:
            # Encontrar punto donde recursivo se vuelve lento
            recursive_data = df[df['Algoritmo'] == 'Recursivo Puro']
            if not recursive_data.empty:
                worst_point = recursive_data.loc[recursive_data['Tiempo'].idxmax()]
                ax1.annotate(
                    f'Recursivo: {worst_point["Tiempo"]:.4f}s\n(Longitud {worst_point["Longitud"]})',
                    xy=(worst_point['Longitud'], worst_point['Tiempo']),
                    xytext=(worst_point['Longitud'] + 1, worst_point['Tiempo'] * 3),
                    arrowprops=dict(arrowstyle='->', color='red', alpha=0.7),
                    fontsize=9, ha='left',
                    bbox=dict(boxstyle="round,pad=0.3", facecolor="lightcoral", alpha=0.7)
                )
        plt.tight_layout()
        return fig

    def create_theoretical_complexity_chart(self):
        """
        Crea gráfico de complejidad teórica con análisis estadístico
        """
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

        n_values = np.arange(1, 16)

        # Complejidades teóricas
        linear = n_values
        quadratic = n_values ** 2
        exponential = 2 ** n_values

        # Gráfico 1: Complejidades teóricas
        ax1.plot(n_values, linear, 'g-', label='O(n) - Ideal', linewidth=3, marker='o', markersize=6)
        ax1.plot(n_values, quadratic, 'b-', label='O(n²) - Prog. Dinámica/Memoización', linewidth=3, marker='s', markersize=6)
        ax1.plot(n_values[:12], exponential[:12], 'r-', label='O(2ⁿ) - Recursivo Puro', linewidth=3, marker='^', markersize=6)

        ax1.set_yscale('log')
        ax1.set_title('📈 Complejidad Temporal Teórica', fontsize=14, fontweight='bold', pad=20)
        ax1.set_xlabel('Tamaño del Problema (n)', fontsize=12)
        ax1.set_ylabel('Número de Operaciones (escala log)', fontsize=12)
        ax1.legend(fontsize=10)
        ax1.grid(True, alpha=0.3)

        # Añadir zona de "inviabilidad práctica"
        ax1.axvspan(12, 15, alpha=0.2, color='red',
                   label='Zona impráctica para O(2ⁿ)')
        ax1.text(13.5, 1000, 'Zona\nImpráctica', ha='center', va='center',
                bbox=dict(boxstyle="round,pad=0.3", facecolor="lightcoral", alpha=0.8),
                fontsize=9, fontweight='bold')

        # Gráfico 2: Análisis de crecimiento relativo
        n_comparison = np.arange(5, 21)
        quad_growth = n_comparison ** 2
        exp_growth = 2 ** n_comparison[:10]  # Solo hasta n=14 para exponencial

        # Normalizar para comparación
        quad_normalized = quad_growth / quad_growth[0]
        exp_normalized = exp_growth / exp_growth[0]

        ax2.plot(n_comparison, quad_normalized, 'b-', label='Crecimiento O(n²)',
                linewidth=3, marker='s', markersize=6)
        ax2.plot(n_comparison[:10], exp_normalized, 'r-', label='Crecimiento O(2ⁿ)',
                linewidth=3, marker='^', markersize=6)

        ax2.set_yscale('log')
        ax2.set_title('📊 Factor de Crecimiento Relativo', fontsize=14, fontweight='bold', pad=20)
        ax2.set_xlabel('Tamaño del Problema (n)', fontsize=12)
        ax2.set_ylabel('Factor de Crecimiento (base = n=5)', fontsize=12)
        ax2.legend(fontsize=10)
        ax2.grid(True, alpha=0.3)

        # Análisis estadístico con scipy
        correlation_quad = stats.pearsonr(n_comparison, quad_normalized)[0]

        ax2.text(0.05, 0.95, f'Correlación O(n²): {correlation_quad:.3f}',
                transform=ax2.transAxes, fontsize=10,
                bbox=dict(boxstyle="round,pad=0.3", facecolor="lightblue", alpha=0.8))

        plt.tight_layout()
        return fig

    def generate_summary_report(self, df):
        """Genera reporte resumido con estadísticas clave"""
        print(f"\n{'='*70}")
        print("📋 REPORTE DE RENDIMIENTO RESUMIDO")
        print(f"{'='*70}")

        if df.empty:
            print("❌ No hay datos para generar el reporte")
            return

        # Estadísticas por algoritmo usando scipy
        print("\n📊 ESTADÍSTICAS DESCRIPTIVAS:")
        print("-" * 50)

        for algorithm in df['Algoritmo'].unique():
            algo_data = df[df['Algoritmo'] == algorithm]['Tiempo']

            # Estadísticas con scipy
            mean_time = np.mean(algo_data)
            std_time = np.std(algo_data, ddof=1)  # Desviación estándar muestral
            median_time = np.median(algo_data)

            print(f"\n🔸 {algorithm}:")
            print(f"   Media: {mean_time:.6f}s")
            print(f"   Mediana: {median_time:.6f}s")
            print(f"   Desv. Est.: {std_time:.6f}s")
            print(f"   Coef. Variación: {(std_time/mean_time)*100:.2f}%")

        # Análisis de escalabilidad
        print(f"\n🚀 ANÁLISIS DE ESCALABILIDAD:")
        print("-" * 50)

        for algorithm in df['Algoritmo'].unique():
            algo_data = df[df['Algoritmo'] == algorithm].sort_values('Longitud')
            if len(algo_data) > 2:
                first_time = algo_data['Tiempo'].iloc[0]
                last_time = algo_data['Tiempo'].iloc[-1]
                first_length = algo_data['Longitud'].iloc[0]
                last_length = algo_data['Longitud'].iloc[-1]

                growth_factor = last_time / first_time
                length_factor = last_length / first_length

                print(f"🔸 {algorithm}:")
                print(f"   Factor de crecimiento: {growth_factor:.2f}x")
                print(f"   Para {length_factor:.1f}x más datos")
                print(f"   Eficiencia: {'🟢 Excelente' if growth_factor < 10 else '🟡 Buena' if growth_factor < 100 else '🔴 Pobre'}")

```

```python

from typing import List, Tuple

# Configuración de estilo optimizada
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['font.size'] = 11

class KnapsackAnalysis:
    """
    Análisis simplificado del problema de la mochila 0/1
    Enfocado en demostración clara y comparación esencial de algoritmos
    """

    def __init__(self):
        self.algorithms = {
            'Recursivo Puro': self._knapsack_recursive,
            'Memoización': self._knapsack_memoization,
            'Prog. Dinámica': self._knapsack_dp,
        }

    def _knapsack_recursive(self, weights: List[int], values: List[int],
                           capacity: int, n: int = None) -> int:
        """Implementación recursiva pura O(2^n)"""
        if n is None:
            n = len(weights)

        if n == 0 or capacity == 0:
            return 0

        if weights[n-1] > capacity:
            return self._knapsack_recursive(weights, values, capacity, n-1)

        include = values[n-1] + self._knapsack_recursive(
            weights, values, capacity - weights[n-1], n-1)
        exclude = self._knapsack_recursive(weights, values, capacity, n-1)

        return max(include, exclude)

    def _knapsack_memoization(self, weights: List[int], values: List[int],
                             capacity: int) -> int:
        """Implementación con memoización O(n*W)"""
        n = len(weights)
        memo = {}

        def helper(i: int, w: int) -> int:
            if i == 0 or w == 0:
                return 0

            if (i, w) in memo:
                return memo[(i, w)]

            if weights[i-1] > w:
                result = helper(i-1, w)
            else:
                include = values[i-1] + helper(i-1, w - weights[i-1])
                exclude = helper(i-1, w)
                result = max(include, exclude)

            memo[(i, w)] = result
            return result

        return helper(n, capacity)

    def _knapsack_dp(self, weights: List[int], values: List[int],
                     capacity: int) -> int:
        """Implementación con programación dinámica O(n*W)"""
        n = len(weights)
        dp = [[0 for _ in range(capacity + 1)] for _ in range(n + 1)]

        for i in range(1, n + 1):
            for w in range(1, capacity + 1):
                if weights[i-1] <= w:
                    include = values[i-1] + dp[i-1][w - weights[i-1]]
                    exclude = dp[i-1][w]
                    dp[i][w] = max(include, exclude)
                else:
                    dp[i][w] = dp[i-1][w]

        return dp[n][capacity]

    def generate_test_data(self, max_items=10):
        """Genera datos de prueba para análisis de rendimiento"""
        random.seed(42)  # Para reproducibilidad
        results = []

        print("🔄 Ejecutando análisis de rendimiento del problema de la mochila...")
        print("-" * 60)

        for n_items in range(1, max_items + 1):
            # Generar caso de prueba más pequeño
            weights = [random.randint(1, 8) for _ in range(n_items)]
            values = [random.randint(5, 15) for _ in range(n_items)]
            capacity = sum(weights) // 3  # Capacidad del 33% del peso total

            print(f"📦 Items: {n_items}, Capacidad: {capacity}")

            for alg_name, algorithm in self.algorithms.items():
                # Limitar recursivo puro para evitar timeouts
                if n_items > 12 and alg_name == 'Recursivo Puro':
                    continue

                try:
                    start_time = time.time()
                    result = algorithm(weights, values, capacity)
                    execution_time = time.time() - start_time

                    results.append({
                        'Items': n_items,
                        'Algoritmo': alg_name,
                        'Tiempo': execution_time,
                        'Resultado': result,
                        'Capacidad': capacity
                    })

                    print(f"  ✓ {alg_name:18s}: {execution_time:.6f}s (Valor={result})")

                except Exception as e:
                    print(f"  ❌ {alg_name:18s}: Error o timeout")

        return pd.DataFrame(results)

    def demonstrate_dp_step_by_step(self, weights=None, values=None, capacity=None, n_items=10):
      """
      Demuestra paso a paso el algoritmo de programación dinámica
      usando exactamente n_items ítems si no se pasan pesos/valores.
      """
      print(f"\n[🔍 DEMOSTRACIÓN DP PASO A PASO - {n_items} ÍTEMS]")

      # 1) Generar datos aleatorios si no se proveen
      if weights is None or values is None or capacity is None:
          random.seed(1)  # reproducible
          weights = [random.randint(1, 8) for _ in range(n_items)]
          values  = [random.randint(5, 15) for _ in range(n_items)]
          capacity = sum(weights) // 2

      # 2) Preparar variables
      n = len(weights)
      print("Pesos:", weights)
      print("Valores:", values)
      print("Capacidad máxima:", capacity)

      # 3) Inicializar tabla DP
      dp = [[0] * (capacity + 1) for _ in range(n + 1)]
      item_labels = [f"E{i}" for i in range(1, n + 1)]
      decisions = []

      # 4) Llenado bottom‑up
      for i in range(1, n + 1):
          for w in range(1, capacity + 1):
              if weights[i - 1] <= w:
                  include = values[i - 1] + dp[i - 1][w - weights[i - 1]]
                  exclude = dp[i - 1][w]
                  if include > exclude:
                      dp[i][w] = include
                      decisions.append((item_labels[i - 1], w, "✔ incluir"))
                  else:
                      dp[i][w] = exclude
                      decisions.append((item_labels[i - 1], w, "✘ excluir"))
              else:
                  dp[i][w] = dp[i - 1][w]
                  decisions.append((item_labels[i - 1], w, "✘ no cabe"))

      # 5) Mostrar decisiones
      print("\nDecisiones tomadas:")
      for item, cap, decision in decisions:
          print(f"Item {item} en capacidad {cap}: {decision}")

      # 6) Mostrar tabla
      self._print_compact_dp_table(dp, item_labels, capacity)

      # 7) Reconstruir solución y devolver
      selected_items = self._reconstruct_solution(dp, weights, values, capacity, item_labels)
      total_value = dp[n][capacity]
      print(f"\n🧠 Elementos seleccionados: {selected_items}")
      print(f"🎯 Valor total óptimo: {total_value}")

      return dp, selected_items




    def _print_compact_dp_table(self, dp, item_labels, capacity):
      """Imprime una tabla DP compacta y clara"""
      n = len(item_labels)

      print("\n        ", end="")
      for w in range(capacity + 1):
          print(f"Cap{w:1d}", end=" ")
      print()
      print("      " + "─" * (5 * (capacity + 1)))

      for i in range(n + 1):
          if i == 0:
              print("  ∅   │ ", end="")
          else:
              print(f"{item_labels[i-1]:<5}│ ", end="")

          for w in range(capacity + 1):
              print(f"{dp[i][w]:3d} ", end=" ")
          print()


    def _reconstruct_solution(self, dp, weights, values, capacity, item_labels):
      """Reconstruye la solución óptima desde la tabla DP"""
      n = len(weights)
      selected_items = []
      w = capacity

      for i in range(n, 0, -1):
          if dp[i][w] != dp[i - 1][w]:
              selected_items.append(item_labels[i - 1])
              w -= weights[i - 1]

      return selected_items[::-1]


    def create_essential_visualizations(self, df):
        """
        Crea solo las visualizaciones esenciales de comparación
        """
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

        # Paleta de colores moderna
        colors = ['#e74c3c', '#3498db', '#2ecc71']

        # Gráfico 1: Comparación temporal (escala logarítmica)
        sns.lineplot(data=df, x='Items', y='Tiempo', hue='Algoritmo',
                     marker='o', linewidth=3, markersize=10, ax=ax1, palette=colors)

        ax1.set_yscale('log')
        ax1.set_title('⏱️ Comparación de Tiempos de Ejecución',
                      fontsize=14, fontweight='bold', pad=20)
        ax1.set_xlabel('Número de Items', fontsize=12)
        ax1.set_ylabel('Tiempo de Ejecución (segundos, escala log)', fontsize=12)
        ax1.grid(True, alpha=0.3)
        ax1.legend(title='Algoritmo', fontsize=11, title_fontsize=12)

        # Agregar anotación para el punto crítico del recursivo
        recursive_data = df[df['Algoritmo'] == 'Recursivo Puro']
        if not recursive_data.empty:
            worst_recursive = recursive_data.loc[recursive_data['Tiempo'].idxmax()]
            ax1.annotate(
                f'Recursivo: {worst_recursive["Tiempo"]:.4f}s\n(Límite práctico n≤10)',
                xy=(worst_recursive['Items'], worst_recursive['Tiempo']),
                xytext=(worst_recursive['Items'] + 1, worst_recursive['Tiempo'] * 5),
                arrowprops=dict(arrowstyle='->', color='red', alpha=0.8, lw=2),
                fontsize=10, ha='left',
                bbox=dict(boxstyle="round,pad=0.3", facecolor="lightcoral", alpha=0.8)
            )

        # Gráfico 2: Comparación de escalabilidad (solo algoritmos eficientes)
        efficient_df = df[df['Algoritmo'] != 'Recursivo Puro']
        if not efficient_df.empty:
            sns.lineplot(data=efficient_df, x='Items', y='Tiempo', hue='Algoritmo',
                        marker='s', linewidth=3, markersize=8, ax=ax2, palette=colors[1:])

            # Agregar líneas de tendencia con regresión
            for algorithm in efficient_df['Algoritmo'].unique():
                algo_data = efficient_df[efficient_df['Algoritmo'] == algorithm]
                if len(algo_data) > 2:
                    # Calcular regresión lineal
                    items = algo_data['Items'].values
                    times = algo_data['Tiempo'].values
                    slope, intercept, r_value, p_value, std_err = stats.linregress(items, times)

                    # Línea de tendencia
                    x_trend = np.linspace(items.min(), items.max() + 2, 100)
                    y_trend = slope * x_trend + intercept
                    ax2.plot(x_trend, y_trend, '--', alpha=0.7, linewidth=2)

                    # Mostrar R² en la gráfica
                    ax2.text(0.02, 0.98 - (list(efficient_df['Algoritmo'].unique()).index(algorithm) * 0.1),
                            f'{algorithm}: R² = {r_value**2:.3f}',
                            transform=ax2.transAxes, fontsize=10,
                            bbox=dict(boxstyle="round,pad=0.2", facecolor="lightblue", alpha=0.7))

        ax2.set_title('🚀 Escalabilidad: Algoritmos Eficientes',
                      fontsize=14, fontweight='bold', pad=20)
        ax2.set_xlabel('Número de Items', fontsize=12)
        ax2.set_ylabel('Tiempo (segundos)', fontsize=12)
        ax2.legend(title='Algoritmo', fontsize=11, title_fontsize=12)
        ax2.grid(True, alpha=0.3)

        plt.tight_layout()
        return fig

    def create_theoretical_complexity_chart(self):
        """
        Crea gráfico complejidad teórica
        """
        fig, ax = plt.subplots(1, 1, figsize=(12, 8))

        # Datos para análisis teórico
        n_values = np.arange(1, 16)

        # Complejidades teóricas
        exponential = 2 ** n_values
        polynomial_10 = n_values * 10  # O(n*W) con W=10
        polynomial_20 = n_values * 20  # O(n*W) con W=20

        ax.plot(n_values, exponential, 'r-', label='O(2ⁿ) - Recursivo Puro',
                linewidth=4, marker='^', markersize=8)
        ax.plot(n_values, polynomial_10, 'b-', label='O(n×W) - DP (W=10)',
                linewidth=3, marker='s', markersize=6)
        ax.plot(n_values, polynomial_20, 'g-', label='O(n×W) - DP (W=20)',
                linewidth=3, marker='o', markersize=6)

        ax.set_yscale('log')
        ax.set_title('📈 Complejidad Temporal Teórica: Exponencial vs Polinomial',
                      fontsize=16, fontweight='bold', pad=20)
        ax.set_xlabel('Número de Items (n)', fontsize=14)
        ax.set_ylabel('Número de Operaciones (escala logarítmica)', fontsize=14)
        ax.legend(fontsize=12, loc='upper left')
        ax.grid(True, alpha=0.3)

        # Marcar zona impráctica para recursivo
        ax.axvspan(10, 15, alpha=0.2, color='red')
        ax.text(12.5, 1000, 'Zona Impráctica\npara O(2ⁿ)',
                ha='center', va='center', fontsize=12, fontweight='bold',
                bbox=dict(boxstyle="round,pad=0.4", facecolor="lightcoral", alpha=0.9))

        # Añadir puntos de comparación específicos
        ax.axhline(y=1024, color='red', linestyle=':', alpha=0.7)
        ax.text(8, 1500, '2¹⁰ = 1,024 operaciones', fontsize=10,
                bbox=dict(boxstyle="round,pad=0.2", facecolor="lightyellow", alpha=0.8))

        ax.axhline(y=200, color='blue', linestyle=':', alpha=0.7)
        ax.text(3, 300, '10×20 = 200 operaciones', fontsize=10,
                bbox=dict(boxstyle="round,pad=0.2", facecolor="lightcyan", alpha=0.8))

        plt.tight_layout()
        return fig

    def generate_focused_statistical_report(self, df):
        """Genera reporte estadístico enfocado usando scipy"""

        print("📊 REPORTE ESTADÍSTICO ESENCIAL")
        print(f"{'='*80}")

        if df.empty:
            print("❌ No hay datos para análisis")
            return

        # Estadísticas descriptivas básicas
        print("\n📈 ESTADÍSTICAS POR ALGORITMO:")
        print("-" * 50)

        for algorithm in df['Algoritmo'].unique():
            algo_data = df[df['Algoritmo'] == algorithm]['Tiempo']

            mean_time = np.mean(algo_data)
            std_time = np.std(algo_data, ddof=1)
            min_time = np.min(algo_data)
            max_time = np.max(algo_data)

            print(f"\n🔸 {algorithm}:")
            print(f"   Media:          {mean_time:.6f}s")
            print(f"   Desv. Estándar: {std_time:.6f}s")
            print(f"   Rango:          [{min_time:.6f}, {max_time:.6f}]s")

        # Análisis de mejora de rendimiento
        print(f"\n⚡ COMPARACIÓN DE EFICIENCIA:")
        print("-" * 50)

        mean_times = df.groupby('Algoritmo')['Tiempo'].mean()
        if 'Recursivo Puro' in mean_times.index:
            recursive_time = mean_times['Recursivo Puro']

            for algo, time in mean_times.items():
                if algo != 'Recursivo Puro':
                    improvement = recursive_time / time
                    print(f"   {algo:15s}: {improvement:.1f}x más rápido que Recursivo")

        # Análisis de escalabilidad con regresión
        print(f"\n🚀 ANÁLISIS DE ESCALABILIDAD (SCIPY):")
        print("-" * 50)

        for algorithm in df['Algoritmo'].unique():
            algo_data = df[df['Algoritmo'] == algorithm].sort_values('Items')
            if len(algo_data) > 2:
                items = algo_data['Items'].values
                times = algo_data['Tiempo'].values

                slope, intercept, r_value, p_value, std_err = stats.linregress(items, times)

                print(f"\n🔸 {algorithm}:")
                print(f"   Crecimiento:  {slope:.8f} s/item")
                print(f"   R²:           {r_value**2:.4f}")
                print(f"   P-value:      {p_value:.2e}")

                # Predicción para problema más grande
                pred_15 = slope * 15 + intercept
                print(f"   Predicción n=15: {pred_15:.6f}s")

```

```python
import random, time
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 8)
plt.rcParams['font.size'] = 11

class KnapsackAnalysis:
    def __init__(self):
        self.algorithms = {
            'Recursivo Puro': self._knapsack_recursive,
            'Memoización': self._knapsack_memoization,
            'Prog. Dinámica': self._knapsack_dp,
        }

    def _knapsack_recursive(self, weights, values, capacity, n=None):
        if n is None: n = len(weights)
        if n==0 or capacity==0: return 0
        if weights[n-1] > capacity:
            return self._knapsack_recursive(weights, values, capacity, n-1)
        include = values[n-1] + self._knapsack_recursive(weights, values, capacity-weights[n-1], n-1)
        exclude = self._knapsack_recursive(weights, values, capacity, n-1)
        return max(include, exclude)

    def _knapsack_memoization(self, weights, values, capacity):
        n = len(weights); memo={}
        def helper(i,w):
            if i==0 or w==0: return 0
            if (i,w) in memo: return memo[(i,w)]
            if weights[i-1]>w:
                res = helper(i-1,w)
            else:
                res = max(values[i-1]+helper(i-1,w-weights[i-1]), helper(i-1,w))
            memo[(i,w)] = res
            return res
        return helper(n,capacity)

    def _knapsack_dp(self, weights, values, capacity):
        n = len(weights)
        dp = [[0]*(capacity+1) for _ in range(n+1)]
        for i in range(1,n+1):
            for w in range(1,capacity+1):
                if weights[i-1]<=w:
                    dp[i][w] = max(values[i-1]+dp[i-1][w-weights[i-1]], dp[i-1][w])
                else:
                    dp[i][w] = dp[i-1][w]
        return dp[n][capacity]

    def demonstrate_dp_step_by_step(self, weights=None, values=None, capacity=None, n_items=10):
        if weights is None or values is None or capacity is None:
            random.seed(1)
            weights = [random.randint(1,8) for _ in range(n_items)]
            values  = [random.randint(5,15) for _ in range(n_items)]
            capacity = sum(weights)//2
        n = len(weights)
        print(f"\n🔍 DEMO DP con {n} ítems")
        print("Pesos:", weights)
        print("Valores:", values)
        print("Capacidad:", capacity)
        dp = [[0]*(capacity+1) for _ in range(n+1)]
        labels = [f"E{i}" for i in range(1,n+1)]
        for i in range(1,n+1):
            for w in range(1,capacity+1):
                if weights[i-1]<=w:
                    dp[i][w] = max(values[i-1]+dp[i-1][w-weights[i-1]], dp[i-1][w])
                else:
                    dp[i][w] = dp[i-1][w]
        print("Tabla DP final (última fila):", dp[n])
        selected = []
        w = capacity
        for i in range(n,0,-1):
            if dp[i][w]!=dp[i-1][w]:
                selected.append(labels[i-1])
                w -= weights[i-1]
        selected.reverse()
        print("Seleccionados:", selected)
        return dp, selected

def main():
    analyzer = KnapsackAnalysis()
    dp, sel = analyzer.demonstrate_dp_step_by_step(n_items=10)
    print("-> Ítems en demo:", len(dp)-1)

main()
```
