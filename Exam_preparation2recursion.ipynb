```python
'''
Considere el siguiente algoritmo recursivo.
'''

def sumas(x, n, s):
  if n == 0:
    return s
  else:
    return sumas(x, n-1, s+x[n-1])

x = [11,4,9,6,5,7]
# s = 0 ---> primera llamada recursiva
# 0 + 7 = s = 7 ----> segunda llamada recursiva
# 7 + 5 = s = 12 ----> Tercera llamada
# 12 + 6 = s = 18 -----> cuarta llamada
# 18 + 9 = s = 27 -----> quinta llamada
# 27 + 4 = s = 31 ----> sexta llamada
# 31 + 11 = s = 42 ----> Septima


sumas(x, len(x), 0)


```

```python

'''
El siguiente programa recursivo realiza el calculo de la suma de los valores enteros positivos de
n + (n − 2) + (n − 4) + . . . + (n − n):
'''

def suma_series(n):
  if n < 1:
    return 0
  else:
    return n + suma_series(n-2)

suma_series(10)

```
