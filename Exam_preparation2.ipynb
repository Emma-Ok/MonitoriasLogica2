```python
'''
Considere el siguiente algoritmo recursivo.
'''

def sumas(x, n, s):
  if n == 0:
    return s
  else:
    return sumas(x, n-1, s+x[n-1])

x = [11,4,9,6,5,7]
# s
# 0 ---> primera
# 0 + 7 = 7 ----> segunda
# 7 + 5 = 12 ----> Tercera
# 12 + 6 = 18 ---> Cuarta
# 18 + 9 = 27 ----> Quinta
# 27 + 4 = 31 ---> Sexta
# 31 + 11 = 42 ---> Septima y última

# Retorna 42


sumas(x, len(x), 0)


# Si x = [11,4,9,6,5,7], n=len(x), y s inicia en s=0. Determine cuál es la operación que realiza y el resultado que genera la linea 9.
```

```python

'''
El siguiente programa recursivo realiza el calculo de la suma de los valores enteros positivos de
n + (n − 2) + (n − 4) + . . . + (n − n):
'''

def suma_series(n):
  if n < 1:
    return 0
  else:
    return n + suma_series(n-2)

suma_series(10)

```
