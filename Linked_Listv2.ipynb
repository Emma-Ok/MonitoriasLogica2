# Listas enlazadas ó ligadas

Un lista ligada es una estructura de datos lineal fundamental en la que los elementos, llamados nodos, se almacenan de forma no contigua. Cada nodo contiene dos partes.

Data. "Valor que tiene representación en el problema"
<br>
Pointer. "Referencia o enlace hacia otro nodo de la secuencia"
<br>

------

***Historia.***
Las listas ligadas fueron Desarrolladas en 1955 por Santiago Shaw y herbert Fazzini cómo la principal estructura de datos para su "Lenguaje de Procesamiento de la Información".
Constituyeron una forma dinámica de gestionar la asignación de memoria y la gestión de datos en los inicios de la informática. A diferencia de las matrices, las listas enlazadas permiten inserciones y supresiones eficientes sin reorganizar toda la estructura de datos, aunque acceder a un elemento arbitrario es
𝑂( 𝑛)
O(n) porque hay que recorrer la lista.

----
Ejemplos actuales de la vida real:

- Sistemas operativos Programación: Las listas enlazadas circulares son una opción natural para la programación round-robin, en la que los procesos se gestionan en una cola circular.
- Historial del navegador y navegación: Las listas doblemente enlazadas permiten navegar hacia delante y hacia atrás de forma eficiente.
- Reproductores de música/listas de reproducción: Las listas enlazadas circulares son útiles cuando la lista de reproducción debe repetirse sin fin.
- Implementación de estructuras de datos avanzadas: Muchas estructuras de datos de alto nivel, como las cachés LRU o ciertas representaciones gráficas (listas de adyacencia), se basan en conceptos de listas enlazadas.

# Caso de uso restaurante.

```python
import random

class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.siguiente = None

class LSL:
    def __init__(self):
        self.cabecera = None

    def insertar(self, valor):
        nuevo_nodo = Nodo(valor)
        if self.cabecera is None:
            self.cabecera = nuevo_nodo
        else:
            nodo_actual = self.cabecera
            while nodo_actual.siguiente:
                nodo_actual = nodo_actual.siguiente
            nodo_actual.siguiente = nuevo_nodo

    def imprimir_facturas(self):
        if self.cabecera is None:
            print("No hay facturas registradas.")
        else:
            nodo_actual = self.cabecera
            while nodo_actual:
                factura = nodo_actual.valor
                print(f"Factura #{factura['id_factura']} - Cliente: {factura['cliente']}")
                print("Items:")
                for item, precio in factura["items"]:
                    print(f"  - {item}: ${precio}")
                print(f"Total: ${factura['total']}")
                print("-" * 40)
                nodo_actual = nodo_actual.siguiente

    def total_vendido(self):
      '''
      Calcula el total vendido en el día.
      '''
      total = 0
      nodo_actual = self.cabecera
      while nodo_actual:
          total += factura != nodo_actual.valor["total"]
          nodo_actual = nodo_actual.siguiente
      return total

    def factura_mas_cara(self):
      '''
      Busca la factura más cara del día.
      '''
      if self.cabecera is None:
          return None
      mayor = self.cabecera.valor
      nodo_actual = self.cabecera.siguiente
      while nodo_actual:
          if nodo_actual.valor["total"] > mayor["total"]:
              mayor = nodo_actual.valor
          nodo_actual = nodo_actual.siguiente
      return mayor

    def buscar_factura_cliente(self, nombre_cliente):
      '''
      Busca facturas de un cliente en particular.
      '''
      resultados = []
      nodo_actual = self.cabecera
      while nodo_actual:
          if nodo_actual.valor["cliente"].lower() == nombre_cliente.lower():
              resultados.append(nodo_actual.valor)
          nodo_actual = nodo_actual.siguiente
      return resultados
```

```python
# ==========================
# INGRESO MANUAL DE FACTURAS
# ==========================
def ingresar_factura(id_factura, clientes, menu):
    print(f"\n--- Crear Factura #{id_factura} ---")

    # Cliente
    print("Clientes disponibles:")
    for i, c in enumerate(clientes, start=1):
        print(f"{i}. {c}")
    opcion_cliente = input("Seleccione cliente por número o escriba uno nuevo: ").strip()

    if opcion_cliente.isdigit() and 1 <= int(opcion_cliente) <= len(clientes):
        cliente = clientes[int(opcion_cliente)-1]
    else:
        cliente = opcion_cliente
        clientes.append(cliente)

    # Items
    items_seleccionados = []
    while True:
        print("\nMenú del restaurante:")
        for i, (item, precio) in enumerate(menu, start=1):
            print(f"{i}. {item} - ${precio}")

        opcion = input("Seleccione un número de item (o 'fin' para terminar): ")
        if opcion.lower() == "fin":
            break
        if opcion.isdigit() and 1 <= int(opcion) <= len(menu):
            item, precio = menu[int(opcion)-1]
            items_seleccionados.append((item, precio))
        else:
            print("⚠ Opción inválida")

    # Calcular total
    total = sum(precio for _, precio in items_seleccionados)

    return {
        "id_factura": id_factura,
        "cliente": cliente,
        "items": items_seleccionados,
        "total": total
    }
```

# LDL LISTA DOBLEMENTE LIGADA

```python
class Nodo:
  def __init__(self, valor):
    self.valor = valor
    self.siguiente = None
    self.anterior = None

class LDL:
  def __init__(self):
    self.cabecera = None
    self.fin = None

  def insertar_al_inicio(self, valor):
    nuevo_nodo = Nodo(valor)
    nuevo_nodo.siguiente = self.cabecera # Puntero siguiente
    nuevo_nodo.anterior = None # puntero anterior
    if self.cabecera is not None: # Primera pregunta la cabecera es null?
      self.cabecera.anterior = nuevo_nodo # En caso de que así sea entonces vamos a crear el nuevo nodo
    else:
      self.fin = nuevo_nodo
    self.cabecera = nuevo_nodo # se actualiza el puntero cabecera

  def insertar_al_final(self, valor):
    nuevo_nodo = Nodo(valor) # creamos el nodo 1ro
    nuevo_nodo.siguiente = None # Puntero siguiente
    nodo_actual = self.cabecera
    if self.cabecera is None: #
      nuevo_nodo.anterior = None
      self.cabecera = nuevo_nodo
      return
    while nodo_actual.siguiente != None:
      nodo_actual = nodo_actual.siguiente # En caso de ser verdad lo almacena en el siguiente
    nodo_actual.siguiente = nuevo_nodo
    nuevo_nodo.anterior = nodo_actual # Actualizar la lista

  def insertar_cualquier(self, valor, nodo_referencia):
    if nodo_referencia is None:
      print("El nodo de referencia no esta en la LDL")
      return
    nuevo_nodo = Nodo(valor)
    nuevo_nodo.siguiente = nodo_referencia.siguiente
    nodo_referencia.siguiente = nuevo_nodo
    nuevo_nodo.anterior = nodo_referencia
    if nuevo_nodo.siguiente is not None:
      nuevo_nodo.siguiente.anterior = nuevo_nodo

  def intercambia_consecutivos(self):
      """Intercambia los nodos consecutivos en pares dentro de la lista."""
      if not self.cabecera or not self.cabecera.siguiente:
        return

      nodo_actual = self.cabecera
      self.cabecera = nodo_actual.siguiente # La cabecera seria el segundo nodo

      while nodo_actual and nodo_actual.siguiente:
        siguiente_nodo = nodo_actual.siguiente # Obtener el segundo nodo del par
        # Ajustar las punteros para intercambiar los nodos
        nodo_actual.siguiente = siguiente_nodo.siguiente # el primer nodo apunta al siguiente despues del par
        siguiente_nodo.anterior = nodo_actual.anterior # El segundo nodo apunta al anterior del primer nodo
        if siguiente_nodo.siguiente:
            siguiente_nodo.siguiente.anterior = nodo_actual # Lo conectamos al primer nodo
        if nodo_actual.anterior:
            nodo_actual.anterior.siguiente = siguiente_nodo # Conectamos el anterior al segundo
        siguiente_nodo.siguiente = nodo_actual # El segundo nodo ahora apunta al primero
        nodo_actual.anterior = siguiente_nodo # El primer nodo apunta al segundo como su anterior
        nodo_actual = nodo_actual.siguiente # Avanzamos al siguiente par de nodos

  def imprimir(self):
    nodo_actual = self.cabecera
    while nodo_actual != None:
      print(nodo_actual.valor, end = "<-->")
      nodo_actual = nodo_actual.siguiente # imprimir cada valor
    print("None")

  def eliminar(self, valor):
        if self.cabecera is None:
            print("La lista esta vacia")
        else:
            nodo_actual = self.cabecera
            while nodo_actual is not None:
                if nodo_actual.valor == valor:
                    break
                nodo_actual = nodo_actual.siguiente

            if nodo_actual is None:
                print("No se encontro el valor en la lista")
                return

            if nodo_actual == self.cabecera:
                self.cabecera = self.cabecera.siguiente
                self.cabecera.anterior = None
                nodo_actual.siguiente = None
            elif nodo_actual == self.fin:
                self.fin = self.fin.anterior
                self.fin.siguiente = None
                nodo_actual.anterior = None
            else:
                nodo_anterior = nodo_actual.anterior
                nodo_siguiente = nodo_actual.siguiente

                nodo_anterior.siguiente = nodo_siguiente
                nodo_siguiente.anterior = nodo_anterior

  def busqueda_binaria(self, valor):
        izquierda = self.cabecera
        derecha = None
        while izquierda != derecha:
            H = self.nodo_medio(izquierda, derecha)
            if H is None:
                return None
            if H.valor == valor:
                return H
            elif H.valor < valor:
                izquierda = H.siguiente
            else:
                derecha = H
        return None

  def nodo_medio(self, cab, fin):
        if cab is None:
            return None
        F = cab
        L = cab.siguiente
        while L != fin:
            L = L.siguiente
            if L != fin:
                L = L.siguiente
                F = F.siguiente
        return F
```

```python
# Prueba del código
lista = LDL()
valores = [1, 2, 3, 4, 5, 6, 7]  # Lista con número impar de elementos
for v in valores:
    lista.insertar_al_final(v)

print("Lista original:")
lista.imprimir()

lista.intercambia_consecutivos()

print("\nLista después del intercambio de pares:")
lista.imprimir()
```

```python
# Clase para representar un término del polinomio
class PolyNodo:
    def __init__(self, coeff, exp):
        self.coeff = coeff      # Coeficiente del término
        self.exp = exp          # Exponente del término
        self.siguiente = None   # Puntero al siguiente nodo
        self.anterior = None    # Puntero al nodo anterior

# Clase para representar un polinomio usando una lista doblemente ligada
class PolyLDL:
    def __init__(self):
        self.cabecera = None    # Primer nodo de la lista
        self.fin = None         # Último nodo de la lista

    def insertar_termino(self, coeff, exp):
        """
        Inserta o actualiza un término en el polinomio.
        La lista se mantiene ordenada de mayor a menor exponente.
        Si ya existe un término con el mismo exponente, se suman los coeficientes.
        Si el coeficiente resultante es 0, se elimina ese término.
        """
        if coeff == 0:
            return

        nuevo = PolyNodo(coeff, exp)

        # Si la lista está vacía, se inserta el primer nodo.
        if self.cabecera is None:
            self.cabecera = self.fin = nuevo
            return

        # Si el exponente del nuevo término es mayor que el de la cabecera, se inserta al inicio.
        if exp > self.cabecera.exp:
            nuevo.siguiente = self.cabecera
            self.cabecera.anterior = nuevo
            self.cabecera = nuevo
            return

        # Buscar la posición para insertar o actualizar un término existente.
        actual = self.cabecera
        while actual is not None:
            # Si se encuentra el mismo exponente, se suman los coeficientes.
            if actual.exp == exp:
                actual.coeff += coeff
                # Si el coeficiente se vuelve cero, se elimina el nodo.
                if actual.coeff == 0:
                    self.eliminar_nodo(actual)
                return
            # Si se encuentra un nodo con menor exponente, se inserta antes.
            elif actual.exp < exp:
                nuevo.siguiente = actual
                nuevo.anterior = actual.anterior
                if actual.anterior:
                    actual.anterior.siguiente = nuevo
                actual.anterior = nuevo
                # Si se insertó al inicio, actualizar la cabecera.
                if nuevo.anterior is None:
                    self.cabecera = nuevo
                return
            # Si no se cumple nada, avanzamos.
            if actual.siguiente is None:
                break
            actual = actual.siguiente

        # Si llegamos al final, insertamos al final.
        actual.siguiente = nuevo
        nuevo.anterior = actual
        self.fin = nuevo

    def eliminar_nodo(self, nodo):
        """Elimina un nodo específico de la lista."""
        if nodo == self.cabecera:
            self.cabecera = nodo.siguiente
            if self.cabecera:
                self.cabecera.anterior = None
        elif nodo == self.fin:
            self.fin = nodo.anterior
            if self.fin:
                self.fin.siguiente = None
        else:
            nodo.anterior.siguiente = nodo.siguiente
            nodo.siguiente.anterior = nodo.anterior

    def imprimir(self):
        """Imprime el polinomio en formato legible."""
        nodo = self.cabecera
        polinomio_str = ""
        while nodo is not None:
            # Formateamos cada término
            termino = f"{nodo.coeff}x^{nodo.exp}" if nodo.exp != 0 else f"{nodo.coeff}"
            if polinomio_str:
                polinomio_str += " + " + termino
            else:
                polinomio_str = termino
            nodo = nodo.siguiente
        print(polinomio_str if polinomio_str else "0")

    def copiar(self):
        """Crea y retorna una copia del polinomio actual."""
        copia = PolyLDL()
        actual = self.cabecera
        while actual:
            copia.insertar_termino(actual.coeff, actual.exp)
            actual = actual.siguiente
        return copia

    @staticmethod
    def sumar(p1, p2):
        """
        Suma dos polinomios (p1 y p2) y retorna un nuevo polinomio resultado.
        Se recorre cada polinomio e inserta sus términos en el resultado.
        La función 'insertar_termino' se encarga de combinar términos con el mismo exponente.
        """
        resultado = PolyLDL()
        actual = p1.cabecera
        while actual:
          resultado.insertar_termino(actual.coeff, actual.exp)
          actual = actual.siguiente

        actual = p2.cabecera
        while actual:
          resultado.insertar_termino(actual.coeff, actual.exp)
          actual = actual.siguiente

        return resultado




    @staticmethod
    def multiplicar(p1, p2):
        """
        Multiplica dos polinomios (p1 y p2) y retorna un nuevo polinomio resultado.
        Para cada término de p1 se multiplica por cada término de p2 y se inserta en el resultado.
        La función 'insertar_termino' se encarga de sumar coeficientes para términos con mismo exponente.
        """
        resultado = PolyLDL()
        actual1 = p1.cabecera
        while actual1:
          actual2 = p2.cabecera
          while actual2:
            nuevo_coeff = actual1.coeff * actual2.coeff
            nuevo_exp = actual1.exp + actual2.exp
            resultado.insertar_termino(nuevo_coeff, nuevo_exp)
            actual2 = actual2.siguiente
          actual1 = actual1.siguiente
        return resultado


```

```python
# Ejemplo de uso para probar la suma y multiplicación de polinomios

# Creamos dos polinomios de ejemplo.
# Por ejemplo, representemos:
# p1(x) = 3x^4 + 2x^2 + 5
# p2(x) = 4x^3 + x^2 + 2

p1 = PolyLDL()
p1.insertar_termino(3, 4)
p1.insertar_termino(2, 2)
p1.insertar_termino(5, 0)

p2 = PolyLDL()
p2.insertar_termino(4, 3)
p2.insertar_termino(1, 2)
p2.insertar_termino(2, 0)

print("Polinomio p1:")
p1.imprimir()  # Debería mostrar: 3x^4 + 2x^2 + 5

print("\nPolinomio p2:")
p2.imprimir()  # Debería mostrar: 4x^3 + 1x^2 + 2

# Sumar los polinomios
suma = PolyLDL.sumar(p1, p2)
print("\nSuma de p1 y p2:")
suma.imprimir()  # Resultado esperado: 3x^4 + 4x^3 + 3x^2 + 7

# Multiplicar los polinomios
multiplicacion = PolyLDL.multiplicar(p1, p2)
print("\nMultiplicación de p1 y p2:")
multiplicacion.imprimir()
# El resultado se obtiene multiplicando cada término:
# 3x^4 * 4x^3 = 12x^7
# 3x^4 * 1x^2 = 3x^6
# 3x^4 * 2 = 6x^4
# 2x^2 * 4x^3 = 8x^5
# 2x^2 * 1x^2 = 2x^4
# 2x^2 * 2 = 4x^2
# 5 * 4x^3 = 20x^3
# 5 * 1x^2 = 5x^2
# 5 * 2 = 10
# Luego se suman términos semejantes.
```
