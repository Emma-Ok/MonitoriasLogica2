# Queues.

En programación, una cola es una estructura de datos que almacena elementos en orden, siguiendo el principio FIFO (first-in, first-out). Esto significa que el primer elemento que se agrega a la cola es el primero que se elimina.

----
*Usos de las colas*
* En sistemas operativos, las colas controlan los procesos dentro de la computadora.
* En aplicaciones que procesan archivos, las colas aseguran que el sistema funcione de manera eficiente.
* En arquitecturas de microservicios y sin servidor, las colas de mensajes almacenan mensajes hasta que se procesan y eliminan.

# Cola Circula CC

```python
class CC:

  def __init__(self, n):
    self.capacidad = n
    self.V = [None]*n # Almacenar los elementos de la cola
    self.primero = self.final = -1 # iniciamos el primero y ultimo con -1

  def ColaVacia(self):
    return self.primero == -1

  def ColaLlena(self):
    return self.primero == (self.final + 1) % self.capacidad

  def EncolarCC(self, valor):
    if self.ColaLlena():
      print("La cola esta llena.")
      return
    else:
      if self.ColaVacia():
        self.primero = 0 # Inserta en la posición 0
      self.final = (self.final + 1) % self.capacidad # Avanzamos con self.final = Final = (Final+ 1) % n
      self.V[self.final] = valor

  def DesencolarCC(self):
    if self.ColaVacia():
      print("La cola esta vacía.")
      return None
    valor_eliminar = self.V[self.primero] # Eliminar el primer valor
    if self.primero == self.final:
      self.primero = self.final = -1 # Retornan a sus condiciones iniciales
    else:
      self.primero = (self.primero + 1) % self.capacidad # Primero = ( primero + 1 ) % n
      self.V[self.primero] = None
      return valor_eliminar

  def recorrer_atras(self):
    if self.ColaVacia():
      print("La cola está vacía")
      return

    p = self.final

    while True:
        print(self.V[p])

        if p == self.primero:
            break

        p = ( p + self.capacidad - 1 ) % self.capacidad

  def MostrarCC(self): # Visualizar el proceso de encolar
    if self.ColaVacia():
      print("La cola esta vacia.")
    else:
      Vector = []
      print("La CC es la siguiente: \n")
      for x in self.V:
        Vector.append(x)
      print(Vector)

  def MostrarCCDesencolar(self): # Visualizar el proceso de desencolar
    if self.ColaVacia():
      print("La CC está vacía")
      return None
    else:
      Vector = []
      i = self.primero
      while True:
        Vector.append(self.V[i])
        if i == self.final:
          break
        i = (i + 1) % self.capacidad # contador para colas
      print(Vector)
```

*Ventajas y desventajas de las Colas No Circulares*:
 * Son recomendables para tareas de insertar y eliminar valores de una forma mas eficiente que las colas circulares.
 * No es necesario usar liıneas de codigo adicionales para definir si Primero = Final.
 * Desde el punto de vista de implementacion en lenguajes de programacion, las colas no circulares son mas faciles de programar por su simplicidad.
 * En el proceso de encolar, exige mas espacio de memoria para almacenar los punteros Primero y Final.
 * Para las operaciones de busqueda de elementos en media de una lista, las colas no circulares no son recomendables debido a su modo de uso y el recorrido de elementos que deberıa realizar para encontrar un elemento.

 ----


 *Ventajas y desventajas de las Colas Circulares*
 * Son recomendables para realizar procesos de busqueda en medio de una lista debido a su flexibilidad para usar los punteros primero y final.
 * Ahorra consumo de memoria puesto que almacena un puntero con respecto al elemento siguiente de la lista.
 * Desafortunadamente, las colas circulares son ineficientes para tareas de insertar o eliminar
 elementos debido al recorrido que deben realizar en la lista. No obstante, si se realiza tareas de actualizacion de valores, puede llegar a ser mas eficientes que las colas no circulares.
 * El manejo de los punteros primero y final requieren de lıneas de codigo adicional, lo cual puede aumentar el costo computacional.

# Arreglo de 2 colas. A2CC


```python
class A2CC: # Cola Circular

  def __init__(self, n):
    m = n // 2 # Dividir en 2 partes
    self.total = n # tamaño total
    self.capacidad = m # Tamaño de cada parte
    self.V = [None]*self.total
    self.P1 = -1
    self.F1 = self.capacidad - 1
    self.P2 = self.capacidad - 1
    self.F2 = self.total - 1

  def ColaLlena(self, puntero_inicial, puntero_final):
    # puntero_inicial = puntero para encolar
    # puntero_final = puntero para desencolar

    # Primera Parte pos 0 hasta pos m - 1
    if puntero_final == (self.capacidad - 1):
      if puntero_inicial == puntero_final:
        return True

    # Segunda Parte pos m hasta pos n - 1
    if puntero_final == (self.total - 1):
      if puntero_inicial == puntero_final:
        return True

  def ColaVacia(self, puntero_inicial, puntero_final):
    # Primera Parte pos 0 hasta pos m - 1
    if puntero_inicial == -1 and puntero_final == self.capacidad - 1:
      return True

    # Segunda Parte pos m hasta pos n - 1
    if puntero_inicial == self.capacidad - 1 and puntero_final == self.total - 1:
      return True

  def Encolar_1(self, valor):
    if self.ColaLlena(self.P1, self.F1) != True:
      self.P1 = (self.P1 + 1) % self.capacidad # P1 = (P1 + 1 ) % m
      self.V[self.P1] = valor
    else:
      print("La cola 1 esta llena")

  def Encolar_2(self, valor):
    if self.ColaLlena(self.P2, self.F2) != True:
      self.P2 = ((self.P2 - self.capacidad + 1) % (self.total - self.capacidad)) + self.capacidad
      # P2 = ((P2 - m + 1) % (n - m)) + m
      self.V[self.P2] = valor
    else:
      print("La cola 2 esta llena")

  def Mostrar(self):
    if self.ColaVacia(self.P1, self.F1) == True:
      return None
    if self.ColaVacia(self.P2, self.F2) == True:
      return None
    print(self.V)
```

# Stack.

Una pila en programación es una estructura de datos abstracta que funciona bajo el principio LIFO ("Last In, First Out")

----


*Las pilas se emplean en múltiples soluciones y algoritmos, entre ellos*

* Evaluación de expresiones: Por ejemplo, para convertir o evaluar expresiones infijas, prefijas o postfijas. Las calculadoras que usan notación polaca inversa (RPN) utilizan pilas para almacenar operandos y aplicar operadores en el orden correcto.

* Recursividad y llamadas a funciones: Cada vez que se llama a una función, se guarda información (variables locales, dirección de retorno) en la pila de llamadas, lo que permite que, al terminar la función, se retome la ejecución en el punto correcto.

* Deshacer/Rehacer (Undo/Redo): En editores de texto o gráficos, se utiliza una pila para almacenar acciones recientes. Al "deshacer", se saca la última acción de la pila.

* Backtracking: Algoritmos de búsqueda en profundidad (DFS) o resolución de laberintos usan pilas para recordar puntos de decisión y retroceder cuando se llega a un callejón sin salida.

* Navegación web: Los botones "atrás" y "adelante" en los navegadores web se basan en pilas para manejar el historial de navegación.


```python
class Pila:

  def __init__(self, n):
    self.V = [None]*n
    self.cima = -1 # Puntero

  def PilaVacia(self):
    if self.cima == -1: # cima esta al inicio del arreglo o lista
      return True

  def PilaLlena(self):
    if self.cima == len(self.V) - 1: # cima esta al final del arreglo o lista
      return True

  def Apilar(self, valor):
    if self.PilaLlena() == True:
      print("La pila está llena")
      return
    else:
      self.cima += 1 # Contador Incremental
      self.V[self.cima] = valor # Apilar el valor

  def Desapilar(self):
    if self.PilaVacia() == True:
      print("La pila está vacía")
      return
    else:
      Valor_Eliminar = self.V.pop(self.cima)
      self.cima -= 1
      return Valor_Eliminar

  def desapilar_y_almacenar(self):
    lista_nueva = []
    while not self.PilaVacia():
        valor = self.Desapilar()
        if valor is not None:
          lista_nueva.append(valor)

    return lista_nueva

  def DesapilarNuevo(self, c):
    if (self.cima - c) >= 0:
      self.cima -= c # cima = cima - c
      valor_eliminar = self.V.pop(self.cima)
      return valor_eliminar
    else:
      print("Solo se elimina hasta la posición c = {:.0f}".format(c))

  def Mostrar(self):
    print(self.V)

  def detectar_secuencia(self, secuencia):
    '''
    dada una pila y una secuencia ingresada crea una función que detecte si una secuencia de valores está en la pila
    '''



  def controla_emparejamientos(self, cadena):
    '''
    Utiliza una pila para resolver un problema de emparejamiento de caracteres.
    Debe leer una cadena e indicar si está bien o mal emparejada.
    '''
```

```python

# Prueba de controla_emparejamientos
print("==== PRUEBA DE CONTROL DE EMPAREJAMIENTOS ====")

# Caso 1: Cadena bien emparejada con caracteres normales
pila1 = Pila(50)  # Creamos una pila con capacidad para 50 elementos
cadena1 = "Este es un (ejemplo) de [texto] con {llaves} bien emparejadas."
resultado1 = pila1.controla_emparejamientos(cadena1)
print(f"Cadena: {cadena1}")
print(f"¿Está bien emparejada? {resultado1}")
print(f"Caracteres normales: {''.join(pila1.pila_caracteres_normales)}")
print()

# Caso 2: Cadena mal emparejada
pila2 = Pila(50)
cadena2 = "Este [texto tiene (un error] de emparejamiento)."
resultado2 = pila2.controla_emparejamientos(cadena2)
print(f"Cadena: {cadena2}")
print(f"¿Está bien emparejada? {resultado2}")
print(f"Caracteres normales: {''.join(pila2.pila_caracteres_normales)}")
print()

# Caso 3: Solo caracteres normales
pila3 = Pila(50)
cadena3 = "Esta cadena solo tiene caracteres normales."
resultado3 = pila3.controla_emparejamientos(cadena3)
print(f"Cadena: {cadena3}")
print(f"¿Está bien emparejada? {resultado3}")
print(f"Caracteres normales: {''.join(pila3.pila_caracteres_normales)}")
print()

# Prueba de detectar_secuencia
print("==== PRUEBA DE DETECCIÓN DE SECUENCIA ====")

# Creamos una pila y agregamos elementos
pila4 = Pila(10)
elementos = [1, 2, 3, 4, 5, 6, 7]
for elem in elementos:
    pila4.Apilar(elem)

# Guardamos el estado actual de la pila para mostrar
elementos_pila = [pila4.V[i] for i in range(pila4.cima + 1)]

# Probamos diferentes secuencias
secuencias_prueba = [
    [3, 4, 5],  # Secuencia que sí está en la pila
    [1, 2, 3],  # Secuencia que sí está en la pila (al inicio)
    [5, 6, 7],  # Secuencia que sí está en la pila (al final)
    [1, 3, 5],  # Secuencia que no está en la pila
    [8, 9, 10]  # Secuencia que no está en la pila
]

for secuencia in secuencias_prueba:
    resultado = pila4.detectar_secuencia(secuencia)
    print(f"Elementos en la pila: {elementos_pila}")
    print(f"Secuencia buscada: {secuencia}")
    print(f"¿Se encontró la secuencia? {resultado}")
    print()

# Prueba extra: desapilar_y_almacenar
print("==== PRUEBA DE DESAPILAR Y ALMACENAR ====")
pila5 = Pila(10)
for i in range(1, 6):
    pila5.Apilar(i)

print("Elementos en la pila antes de desapilar:")
pila5.Mostrar()

lista_resultado = pila5.desapilar_y_almacenar()

print("Elementos recuperados y almacenados en lista:")
print(lista_resultado)

print("Estado de la pila después de desapilar:")
pila5.Mostrar()
```
