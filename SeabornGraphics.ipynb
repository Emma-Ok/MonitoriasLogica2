# Seaborn library

Seaborn es una biblioteca para hacer gráficos estadísticos en Python. Se basa en matplotlib y se integra estrechamente con las estructuras de datos de pandas.

Seaborn le ayuda a explorar y comprender sus datos. Sus funciones de trazado operan en marcos de datos y matrices que contienen conjuntos de datos completos y realizan internamente el mapeo semántico y la agregación estadística necesarios para producir gráficos informativos. Su API declarativa orientada a conjuntos de datos le permite centrarse en el significado de los distintos elementos de sus gráficos, en lugar de en los detalles de cómo dibujarlos.

url: https://seaborn.pydata.org/index.html

1.1 ¿Qué es Seaborn?

Seaborn es una librería de visualización de datos basada en matplotlib.
Características principales:
- Interfaz de alto nivel para gráficos estadísticos atractivos
- Integración nativa con pandas DataFrames
- Estilos predefinidos y paletas de colores profesionales
- Funciones especializadas para análisis exploratorio de datos


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
warnings.filterwarnings('ignore')

# Configuración inicial
plt.style.use('default')
sns.set_palette("pastel")
sns.set_style("whitegrid")


# 1.2 Configuración básica
print("\n1.2 Configuración básica")
sns.set_theme()  # Aplicar estilo seaborn por defecto
plt.rcParams['figure.figsize'] = (10, 6)  # Tamaño por defecto

# 1.3 Datos de ejemplo
print("\n1.3 Cargando datos de ejemplo")
# Usaremos datasets incluidos en seaborn
tips = sns.load_dataset('tips')
flights = sns.load_dataset('flights')
iris = sns.load_dataset('iris')
titanic = sns.load_dataset('titanic')

print("Datasets disponibles en seaborn:")
print(sns.get_dataset_names())

print("\nPrimeras filas del dataset 'tips':")
print(tips.head())
```

#2. Gráficos Básicos

```python
# 2.1 Scatter Plot (Gráfico de dispersión)
print("\n2.1 SCATTER PLOTS")
plt.figure(figsize=(15, 5))

# Básico
plt.subplot(1, 3, 1)
sns.scatterplot(data=tips, x='total_bill', y='tip')
plt.title('Scatter Plot Básico')

# Con color por categoría
plt.subplot(1, 3, 2)
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex')
plt.title('Con categoría de color')

# Con tamaño y estilo
plt.subplot(1, 3, 3)
sns.scatterplot(data=tips, x='total_bill', y='tip',
                hue='sex', size='size', style='time')
plt.title('Con múltiples dimensiones')

plt.tight_layout()
plt.show()
```

```python

# 2.2 Line Plot (Gráfico de líneas)
print("\n2.2 LINE PLOTS")
plt.figure(figsize=(12, 4))

plt.subplot(1, 2, 1)
sns.lineplot(data=flights, x='year', y='passengers')
plt.title('Line Plot Básico')

plt.subplot(1, 2, 2)
sns.lineplot(data=flights, x='year', y='passengers', hue='month')
plt.title('Múltiples líneas por categoría')

plt.tight_layout()
plt.show()
```

```python
 #2.3 Bar Plots
print("\n2.3 BAR PLOTS")
plt.figure(figsize=(15, 5))

# Bar plot vertical
plt.subplot(1, 3, 1)
sns.barplot(data=tips, x='day', y='total_bill')
plt.title('Bar Plot Vertical')

# Bar plot horizontal
plt.subplot(1, 3, 2)
sns.barplot(data=tips, x='total_bill', y='day', orient='h')
plt.title('Bar Plot Horizontal')

# Con categorías anidadas
plt.subplot(1, 3, 3)
sns.barplot(data=tips, x='day', y='total_bill', hue='sex')
plt.title('Con categorías anidadas')

plt.tight_layout()
plt.show()
```

```python
# 2.4 Histogramas y Distribuciones
print("\n2.4 HISTOGRAMAS Y DISTRIBUCIONES")
plt.figure(figsize=(12, 8))

# Histograma básico
plt.subplot(2, 3, 1)
sns.histplot(data=tips, x='total_bill')
plt.title('Histograma Básico')

# Histograma con KDE
plt.subplot(2, 3, 2)
sns.histplot(data=tips, x='total_bill', kde=True)
plt.title('Histograma + KDE')

# Distribución por categorías
plt.subplot(2, 3, 3)
sns.histplot(data=tips, x='total_bill', hue='sex', multiple='stack')
plt.title('Por categorías (apilado)')

# Distribución de densidad
plt.subplot(2, 3, 4)
sns.kdeplot(data=tips, x='total_bill')
plt.title('Distribución de densidad (KDE)')

# Múltiples distribuciones
plt.subplot(2, 3, 5)
sns.kdeplot(data=tips, x='total_bill', hue='sex')
plt.title('Múltiples KDE')

# Distribución bidimensional
plt.subplot(2, 3, 6)
sns.kdeplot(data=tips, x='total_bill', y='tip')
plt.title('KDE Bidimensional')

plt.tight_layout()
plt.show()
```

# 3. Gráficas estadísticos

```python
# 3.1 Box Plots
print("\n3.1 BOX PLOTS")
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.boxplot(data=tips, y='total_bill')
plt.title('Box Plot Simple')

plt.subplot(1, 3, 2)
sns.boxplot(data=tips, x='day', y='total_bill')
plt.title('Box Plot por Categoría')

plt.subplot(1, 3, 3)
sns.boxplot(data=tips, x='day', y='total_bill', hue='sex')
plt.title('Box Plot con Subcategorías')

plt.tight_layout()
plt.show()
```

```python
# 3.2 Violin Plots
print("\n3.2 VIOLIN PLOTS")
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
sns.violinplot(data=tips, y='total_bill')
plt.title('Violin Plot Simple')

plt.subplot(1, 3, 2)
sns.violinplot(data=tips, x='day', y='total_bill')
plt.title('Violin Plot por Categoría')

plt.subplot(1, 3, 3)
sns.violinplot(data=tips, x='day', y='total_bill', hue='sex', split=True)
plt.title('Violin Plot Dividido')

plt.tight_layout()
plt.show()
```

```python
# 3.3 Pair Plots
print("\n3.3 PAIR PLOTS")
iris_sample = iris.sample(100)

# Pair plot básico
g = sns.pairplot(iris_sample)
g.fig.suptitle('Pair Plot - Dataset Iris', y=1.02)
plt.show()
```

```python

# Pair plot con categorías
g = sns.pairplot(iris_sample, hue='species')
g.fig.suptitle('Pair Plot con Categorías', y=1.02)
plt.show()
```

```python
# 3.4 Correlation Matrix (Matriz de Correlación)
print("\n3.4 MATRIZ DE CORRELACIÓN")
plt.figure(figsize=(12, 4))

# Calcular correlaciones
tips_numeric = tips.select_dtypes(include=[np.number])
correlation_matrix = tips_numeric.corr()

plt.subplot(1, 2, 1)
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Matriz de Correlación')

plt.subplot(1, 2, 2)
sns.clustermap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Mapa de Correlación Agrupado')

plt.tight_layout()
plt.show()
```

```python
# 4.3 Personalización de Títulos y Etiquetas
print("\n4.3 PERSONALIZACIÓN DE TÍTULOS Y ETIQUETAS")

plt.figure(figsize=(12, 8))

# Gráfico con personalización completa
ax = sns.scatterplot(data=tips, x='total_bill', y='tip',
                     hue='sex', size='size', style='time',
                     sizes=(50, 200))

# Personalizar títulos y etiquetas
plt.title('Relación entre Cuenta Total y Propina\nen Restaurantes',
          fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Cuenta Total ($)', fontsize=12, fontweight='bold')
plt.ylabel('Propina ($)', fontsize=12, fontweight='bold')

# Personalizar leyenda
handles, labels = ax.get_legend_handles_labels()
legend1 = plt.legend(handles[:3], labels[:3], title='Sexo', loc='upper left')
legend2 = plt.legend(handles[3:7], labels[3:7], title='Tamaño Mesa', loc='lower right')
legend3 = plt.legend(handles[8:], labels[8:], title='Momento', loc='upper right')

plt.gca().add_artist(legend1)
plt.gca().add_artist(legend2)

# Personalizar grid
plt.grid(True, alpha=0.3)
sns.despine()

plt.tight_layout()
plt.show()
```

```python
# Crear dataset más complejo
np.random.seed(42)
complex_data = pd.DataFrame({
    'fecha': pd.date_range('2023-01-01', periods=365),
    'ventas': np.random.normal(1000, 200, 365),
    'categoria': np.random.choice(['A', 'B', 'C'], 365),
    'region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], 365),
    'temperatura': np.random.normal(20, 10, 365)
})

# Agregar tendencia estacional
complex_data['ventas'] += np.sin(complex_data.index / 365 * 2 * np.pi) * 100

plt.figure(figsize=(16, 12))

# Serie temporal
plt.subplot(3, 2, 1)
monthly_sales = complex_data.groupby(complex_data['fecha'].dt.month)['ventas'].mean()
sns.lineplot(x=monthly_sales.index, y=monthly_sales.values)
plt.title('Ventas Promedio por Mes')
plt.xlabel('Mes')

# Boxplot por categorías múltiples
plt.subplot(3, 2, 2)
sns.boxplot(data=complex_data, x='categoria', y='ventas', hue='region')
plt.title('Distribución de Ventas por Categoría y Región')

# Scatter con regresión
plt.subplot(3, 2, 3)
sns.scatterplot(data=complex_data, x='temperatura', y='ventas',
                hue='categoria', alpha=0.6)
sns.regplot(data=complex_data, x='temperatura', y='ventas',
            scatter=False, color='red')
plt.title('Relación Temperatura vs Ventas')

# Violin plot
plt.subplot(3, 2, 4)
sns.violinplot(data=complex_data, x='region', y='ventas')
plt.title('Distribución de Ventas por Región')

# Heatmap de correlación
plt.subplot(3, 2, 5)
corr_data = complex_data.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr_data, annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Matriz de Correlación')

# Joint plot
plt.subplot(3, 2, 6)
# Para joint plot, crear por separado
sns.scatterplot(data=complex_data.sample(100), x='temperatura', y='ventas')
plt.title('Muestra: Temperatura vs Ventas')

plt.tight_layout()
plt.show()
```


# CONSEJOS PARA DOMINAR SEABORN:

1. PREPARACIÓN DE DATOS:
   - Usa pandas para limpiar y preparar tus datos
   - Asegúrate de que las columnas categóricas estén bien definidas
   - Maneja valores nulos antes de graficar

2. ELECCIÓN DEL GRÁFICO:
   - Scatter plots: relaciones entre variables continuas
   - Line plots: series temporales o tendencias
   - Bar plots: comparaciones entre categorías
   - Box/Violin plots: distribuciones por grupos
   - Heatmaps: correlaciones y matrices

3. PERSONALIZACIÓN:
   - Usa set_theme() al inicio de tu notebook
   - Define paletas consistentes para tu proyecto
   - Siempre incluye títulos y etiquetas descriptivas
   - Ajusta el tamaño de figura según el contexto

4. PERFORMANCE:
   - Para datasets grandes (>10k filas), considera sampling
   - Usa rasterized=True para scatter plots grandes
   - Evita recrear figuras innecesariamente

5. INTEGRACIÓN:
   - Combina seaborn con matplotlib para mayor control
   - Usa figsize y tight_layout() para mejor presentación
   - Guarda gráficos en alta resolución con dpi=300

6. EXPLORATORY DATA ANALYSIS (EDA):
   - Empieza siempre con pairplot() para overview
   - Usa describe() de pandas antes de graficar
   - Combina múltiples visualizaciones para insights completos
