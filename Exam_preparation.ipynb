```python
class Nodo:
    def __init__(self, valor):
        self.valor = valor
        self.anterior = None
        self.siguiente = None

class LDL:
    def __init__(self):
        self.cabecera = None

    def insertar(self, valor):
        nuevo_nodo = Nodo(valor)
        if self.cabecera is None:
            self.cabecera = nuevo_nodo
            return
        nodo_actual = self.cabecera
        while nodo_actual.siguiente is not None:
            nodo_actual = nodo_actual.siguiente
        nodo_actual.siguiente = nuevo_nodo
        nuevo_nodo.anterior = nodo_actual

    def Prog_LDL(self):
        nodo_actual = self.cabecera
        previo = None
        while nodo_actual is not None:
            previo = nodo_actual.anterior
            nodo_actual.anterior = nodo_actual.siguiente
            nodo_actual.siguiente = previo
            nodo_actual = nodo_actual.anterior
        if previo is not None:
            self.cabecera = previo.anterior
        return self.cabecera

    def mostrar(self):
        nodo_actual = self.cabecera
        while nodo_actual is not None:
            print(nodo_actual.valor, end="<-->")
            nodo_actual = nodo_actual.siguiente
        print("None")

    def Propuesta(self):
        if self.cabecera is None:
            return
        if self.cabecera.siguiente is None:
            self.cabecera = None
            return
        nodo_actual = self.cabecera
        while nodo_actual.siguiente is not None:
            nodo_actual = nodo_actual.siguiente
        nodo_actual.anterior.siguiente = None
        nodo_actual.anterior = None
```

Considere una clase llamada CNC, cuya tarea es realizar operaciones de un Cola No Circular. Dentro de las funciones internas de la clase, la funcion Remover Impares realiza un proceso de eliminacion de los valores impares de la lista y solamente muestra los valores pares de la lista. A partir de lo anterior, disene un algoritmo para remover los valores pares tal que solamente muestre los valores impares de la lista. Posteriormente, determine el costo computacional en el mejor y el peor caso con la nueva funcion disenada.

```python
class CNC:
    def __init__(self, n):
        self.capacidad = n
        self.primero = -1
        self.final = -1
        self.V = [None] * self.capacidad

    def ColaLLena(self):
        return self.final == self.capacidad - 1  # Corregido para que sea válido

    def ColaVacia(self):
        return self.primero == -1 and self.final == -1

    def Encolar(self, valor):
        if self.ColaLLena():
            print("La cola está llena")
            return
        if self.ColaVacia():
            self.primero, self.final = 0, 0
        else:
            self.final += 1
        self.V[self.final] = valor

    def Desencolar(self):
        if self.ColaVacia():
            print("La cola está vacía")
            return None
        valor = self.V[self.primero]
        if self.primero == self.final:
            self.primero, self.final = -1, -1
        else:
            self.primero += 1
        return valor

    def Pico(self):
        if self.ColaVacia():
            print("La cola está vacía")
            return -1
        return self.V[self.primero]

    def Mostrar(self):
        if self.ColaVacia():
            print("La cola está vacía")
            return None
        print("La lista actual es:", end=" ")
        for i in range(self.primero, self.final + 1):
            print(self.V[i], end=" ")
        print()

    def Remover_Impares(self):
        temp = CNC(self.capacidad)
        while not self.ColaVacia():
            x = self.Pico()
            self.Desencolar()
            if x % 2 == 0:  # Si el valor es par
                temp.Encolar(x)

        # Transferir los valores de vuelta a la cola original
        while not temp.ColaVacia():
            self.Encolar(temp.Pico())
            temp.Desencolar()
```
