## ¿Qué es la Programación Dinámica?

La **Programación Dinámica (PD)** es una técnica algorítmica para resolver problemas de optimización dividiéndolos en subproblemas más simples. Se basa en el principio de **optimalidad de Bellman**: una solución óptima al problema contiene soluciones óptimas a los subproblemas.

### Características Principales:

1. **Subestructura Óptima**: El problema se puede dividir en subproblemas más pequeños cuyas soluciones contribuyen a la solución del problema original.

2. **Subproblemas Superpuestos**: Los mismos subproblemas se resuelven múltiples veces. La PD evita recalcularlos almacenando sus resultados.

3. **Memorización (Memoization)**: Técnica top-down que almacena resultados de subproblemas ya resueltos.

4. **Tabulación (Tabulation)**: Técnica bottom-up que construye soluciones desde los casos base hasta el problema completo.

### Programación Dinámica vs Recursión

| Aspecto | Recursión Simple | Programación Dinámica |
|---------|------------------|----------------------|
| **Eficiencia** | Recalcula subproblemas repetidamente | Calcula cada subproblema solo una vez |
| **Complejidad** | Puede ser exponencial O(2ⁿ) | Generalmente polinomial O(n), O(n²) |
| **Memoria** | Usa la pila de llamadas | Usa tabla de memorización |
| **Ejemplo clásico** | Fibonacci recursivo: O(2ⁿ) | Fibonacci con PD: O(n) |

**Ejemplo: Fibonacci**

```python
# Recursión simple - O(2^n)
def fib_recursivo(n):
    if n <= 1:
        return n
    return fib_recursivo(n-1) + fib_recursivo(n-2)

# Programación Dinámica - O(n)
def fib_pd(n):
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]
```

### Pasos para Resolver un Problema con Programación Dinámica

1. **Identificar si el problema tiene subestructura óptima**
   - ¿La solución óptima se puede construir a partir de soluciones óptimas de subproblemas?

2. **Definir el estado**
   - ¿Qué información necesito almacenar para cada subproblema?
   - Ejemplo: `dp[i]` = mejor solución hasta el elemento i

3. **Establecer la relación de recurrencia**
   - ¿Cómo se relaciona `dp[i]` con estados anteriores?
   - Ejemplo: `dp[i] = max(dp[i-1], dp[i-2] + arr[i])`

4. **Determinar los casos base**
   - ¿Cuáles son los valores iniciales conocidos?
   - Ejemplo: `dp[0] = 0`, `dp[1] = arr[0]`

5. **Decidir el orden de cálculo**
   - Bottom-up (tabulación) vs Top-down (memorización)

6. **Optimizar el espacio si es posible**
   - ¿Puedo reducir el espacio de O(n²) a O(n) o O(1)?

### Ejercicio 1: Problema de Selección de Actividades

**Descripción:**

Dado un conjunto de actividades y la hora de inicio y fin de cada actividad, encuentre el número máximo de actividades que puede realizar una sola persona suponiendo que una persona sólo puede trabajar en una única actividad a la vez. La solución solamente se acepta con Programación Dinámica y no se acepta el uso de algoritmos ávaros.

Este problema se llama **problema de selección de actividades** y consiste en seleccionar las actividades no conflictivas que se van a realizar en un plazo determinado, dado un conjunto de actividades marcadas cada una de ellas por una hora de inicio y de finalización.

**Pista 1:**

Entrada: $[(1, 4), (3, 5), (0, 6), (5, 7), (3, 8), (5, 9), (6, 10), (8, 11), (8, 12), (2, 13), (12, 14)]$

Salida: $[(1, 4), (5, 7), (8, 11), (12, 14)]$

**Pista 2:**

Considere la estrategia de solución del problema de la subsecuencia creciente más larga (LIS).

**Pista 3:**

Pueden diseñar el algoritmo mediante el conteo del número máximo de actividades no conflictivas. Es importante que el costo computacional temporal no exceda $\mathcal{O}(n^2)$

````python
def Seleccion_Actividades(aristas):
  """
  Resuelve el problema de selección de actividades usando Programación Dinámica.
  Similar a la estrategia LIS (Longest Increasing Subsequence).

  Tiempo: O(n²)
  Espacio: O(n)
  """
  if not aristas:
    return [], 0

  # Ordenar actividades por tiempo de finalización
  actividades = sorted(aristas, key=lambda x: x[1])
  n = len(actividades)

  # dp[i] = máximo número de actividades no conflictivas hasta la actividad i
  dp = [1] * n

  # parent[i] = índice de la actividad previa en la solución óptima
  parent = [-1] * n

  # Para cada actividad i
  for i in range(1, n):
    # Revisar todas las actividades anteriores j
    for j in range(i):
      # Si la actividad j no se solapa con i (termina antes o cuando empieza i)
      if actividades[j][1] <= actividades[i][0]:
        # Si incluir j mejora la solución para i
        if dp[j] + 1 > dp[i]:
          dp[i] = dp[j] + 1
          parent[i] = j

  # Encontrar el índice con el máximo número de actividades
  max_actividades = max(dp)
  max_idx = dp.index(max_actividades)

  # Reconstruir la solución
  aristas_seleccionadas = []
  idx = max_idx
  while idx != -1:
    aristas_seleccionadas.append(actividades[idx])
    idx = parent[idx]

  aristas_seleccionadas.reverse()

  return aristas_seleccionadas, max_actividades
````

**Explicación del algoritmo:**

Este algoritmo utiliza **Programación Dinámica** basado en la estrategia del problema LIS (Longest Increasing Subsequence):

### Estrategia:

1. **Ordenamiento**: Primero ordenamos las actividades por tiempo de finalización. Esto nos permite procesar las actividades en un orden lógico.

2. **Estado DP**: Definimos `dp[i]` como el número máximo de actividades no conflictivas que podemos realizar considerando las actividades desde 0 hasta i, donde la actividad i está incluida.

3. **Relación de Recurrencia**:
   ```
   dp[i] = max(dp[j] + 1) para todo j < i donde actividades[j][1] <= actividades[i][0]
   ```
   Es decir, buscamos la mejor solución previa donde la actividad j termina antes o cuando comienza la actividad i.

4. **Reconstrucción**: Usamos un arreglo `parent` para rastrear qué actividad previa nos dio la solución óptima, permitiéndonos reconstruir la secuencia completa.

### Pseudocódigo:
```
FUNCIÓN Seleccion_Actividades(aristas):
    ordenar aristas por tiempo de finalización
    inicializar dp[n] = [1, 1, ..., 1]
    inicializar parent[n] = [-1, -1, ..., -1]
    
    PARA i desde 1 hasta n-1:
        PARA j desde 0 hasta i-1:
            SI actividades[j].fin <= actividades[i].inicio:
                SI dp[j] + 1 > dp[i]:
                    dp[i] = dp[j] + 1
                    parent[i] = j
    
    encontrar max_idx donde dp[max_idx] es máximo
    reconstruir solución usando parent
    RETORNAR solución, dp[max_idx]
```

### Costo Computacional:
- **Tiempo**: O(n²) - doble bucle anidado
- **Espacio**: O(n) - arreglos dp y parent
- Cumple con la restricción de no exceder O(n²)

````python
# Prueba 1

aristas = [(1, 4), (3, 5), (0, 6), (5, 7), (3, 8), (5, 9), (6, 10), (8, 11), (8, 12), (2, 13), (12, 14)]
aristas_seleccionadas, conteo_aristas_no_conflictivas = Seleccion_Actividades(aristas)
print("Actividades no conflictivas = ", aristas_seleccionadas)
print("Conteo de actividades no conflictivas = ", conteo_aristas_no_conflictivas)
````

````python
# Prueba 2

aristas = [(0, 3), (0, 4), (0, 5), (1, 3), (1, 4), (1, 5), (2, 3), (2, 4), (2, 5), (3, 4), (3, 4), (3, 5),
           (5, 1), (5, 3), (5, 2), (5, 3), (4, 3), (4, 5)]
aristas_seleccionadas, conteo_aristas_no_conflictivas = Seleccion_Actividades(aristas)
print("Actividades no conflictivas = ", aristas_seleccionadas)
print("Conteo de actividades no conflictivas = ", conteo_aristas_no_conflictivas)
````

### Ejercicio 2: Programación por Intervalos Ponderados

**Descripción**

Dada una lista de trabajos en la que cada trabajo tiene una hora de inicio y de finalización, y un beneficio asociado a él, encuentre un subconjunto de trabajos no solapados de máximo beneficio. Por ejemplo, considere los siguientes trabajos con su hora de inicio, hora de finalización y beneficio asociado.

**Pista:**

El beneficio máximo es $80$, y los trabajos implicados en el beneficio máximo son:

*   $(1, 4, 30)$
*   $(5, 9, 50)$

Al sumar las horas de trabajo, $30 + 50 = 80$, se encuentra el valor máximo beneficio, considerando una lista de trabajos:

* Trabajo 1: $(0, 6, 60)$
* Trabajo 2: $(5, 9, 50)$
* Trabajo 3: $(1, 4, 30)$
* Trabajo 4: $(5, 7, 30)$
* Trabajo 5: $(3, 5, 10)$
* Trabajo 6: $(7, 8, 10)$

Considere la estrategia de solución del problema de la subsecuencia creciente más larga (LIS), usando programación dinámica o backtracking.



````python
def intervalos(horarios):
  """
  Problema de Intervalos Ponderados usando Programación Dinámica.
  Encuentra el subconjunto de trabajos no solapados con máximo beneficio.

  Tiempo: O(n²)
  Espacio: O(n)
  """
  if not horarios:
    return 0, []

  # Ordenar por tiempo de finalización
  trabajos = sorted(horarios, key=lambda x: x[1])
  n = len(trabajos)

  # dp[i] = máximo beneficio considerando trabajos desde 0 hasta i
  dp = [0] * n
  dp[0] = trabajos[0][2]  # El beneficio del primer trabajo

  # Para rastrear qué trabajo incluimos
  incluido = [False] * n
  incluido[0] = True

  # prev_compatible[i] = índice del trabajo compatible previo usado
  prev_compatible = [-1] * n

  # Para cada trabajo i
  for i in range(1, n):
    # Opción 1: No incluir el trabajo i
    beneficio_sin_i = dp[i-1]

    # Opción 2: Incluir el trabajo i
    beneficio_con_i = trabajos[i][2]  # Beneficio del trabajo actual
    mejor_j = -1

    # Buscar el trabajo más reciente j que no se solape con i
    for j in range(i-1, -1, -1):
      if trabajos[j][1] <= trabajos[i][0]:
        beneficio_con_i += dp[j]
        mejor_j = j
        break

    # Elegir la mejor opción
    if beneficio_con_i > beneficio_sin_i:
      dp[i] = beneficio_con_i
      prev_compatible[i] = mejor_j
    else:
      dp[i] = beneficio_sin_i
      prev_compatible[i] = -2  # Marca: no incluimos este trabajo

  # Reconstruir la solución
  maximo_horas = dp[n-1]
  intervalos_seleccionados = []

  # Marcar trabajos incluidos
  idx = n - 1
  while idx >= 0:
    if idx == 0:
      if dp[0] == trabajos[0][2]:
        intervalos_seleccionados.append(trabajos[0])
      break

    # Verificar si el trabajo idx fue incluido
    if prev_compatible[idx] == -2:
      # No fue incluido, retroceder
      idx -= 1
    else:
      # Fue incluido
      # Verificar comparando con la opción de no incluirlo
      beneficio_sin_idx = dp[idx-1] if idx > 0 else 0
      if dp[idx] > beneficio_sin_idx:
        intervalos_seleccionados.append(trabajos[idx])
        idx = prev_compatible[idx]
        if idx == -1:
          break
      else:
        idx -= 1

  intervalos_seleccionados.reverse()

  return maximo_horas, intervalos_seleccionados
````

**Explicación del algoritmo:**

Este algoritmo resuelve el **Problema de Intervalos Ponderados** (Weighted Interval Scheduling) usando Programación Dinámica:

### Estrategia:

1. **Ordenamiento**: Ordenamos los trabajos por tiempo de finalización, igual que en el ejercicio anterior.

2. **Estado DP**: Definimos `dp[i]` como el máximo beneficio que podemos obtener considerando los trabajos desde 0 hasta i.

3. **Relación de Recurrencia**: Para cada trabajo i, tenemos dos opciones:
   ```
   Opción A: No incluir el trabajo i → dp[i] = dp[i-1]
   Opción B: Incluir el trabajo i → dp[i] = beneficio[i] + dp[j]
              donde j es el trabajo más reciente que no se solapa con i
   
   dp[i] = max(Opción A, Opción B)
   ```

4. **Búsqueda del trabajo compatible**: Para cada trabajo i, buscamos hacia atrás el trabajo j más reciente cuyo tiempo de finalización sea ≤ tiempo de inicio de i.

5. **Reconstrucción**: Marcamos qué trabajos fueron incluidos comparando si `dp[i] = dp[previo] + beneficio[i]`.

### Pseudocódigo:
```
FUNCIÓN intervalos(horarios):
    ordenar horarios por tiempo de finalización
    inicializar dp[n]
    dp[0] = beneficio del primer trabajo
    
    PARA i desde 1 hasta n-1:
        beneficio_sin_i = dp[i-1]
        beneficio_con_i = horarios[i].beneficio
        
        PARA j desde i-1 hasta 0 (hacia atrás):
            SI horarios[j].fin <= horarios[i].inicio:
                beneficio_con_i += dp[j]
                ROMPER
        
        dp[i] = max(beneficio_sin_i, beneficio_con_i)
    
    reconstruir solución marcando trabajos incluidos
    RETORNAR dp[n-1], trabajos_seleccionados
```

### Costo Computacional:
- **Tiempo**: O(n²) - para cada trabajo buscamos el compatible previo
- **Espacio**: O(n) - arreglo dp
- Puede optimizarse a O(n log n) usando búsqueda binaria para encontrar trabajos compatibles

### Diferencia con Ejercicio 1:
- **Ejercicio 1**: Maximiza la *cantidad* de actividades
- **Ejercicio 2**: Maximiza el *beneficio* total (ponderado)

````python
# Prueba 1

intervalos_trabajo = [(0, 6, 60), (5, 9, 50), (1, 4, 30), (5, 7, 30), (3, 5, 10),
                      (7, 8, 10)]
maximo_horas, intervalos_seleccionados = intervalos(intervalos_trabajo)
print("Máximo de horas = ", maximo_horas)
print("Intervalos seleccionados = ", intervalos_seleccionados)
````

````python
# Prueba 2

intervalos_trabajo = [(0, 8, 10), (1, 5, 10), (2, 7, 20), (4, 5, 5), (7, 9, 25)]
maximo_horas, intervalos_seleccionados = intervalos(intervalos_trabajo)
print("Máximo de horas = ", maximo_horas)
print("Intervalos seleccionados = ", intervalos_seleccionados)
````

### Ejercicio 3: Conjunto Independiente Máximo - Versión Programación Dinámica

**Descripción:**

Dado un árbol binario, encuentre el tamaño del Conjunto Independiente Máximo (SIM, en inglés) en él. Un conjunto independiente es un conjunto de nodos de un árbol binario, de los cuales no hay dos nodos adyacentes, es decir, no hay ninguna arista que los conecte. El tamaño de un conjunto independiente es el número total de nodos que contiene. Por consiguiente, el problema del conjunto independiente máximo consiste en encontrar un conjunto independiente del mayor tamaño posible para un árbol binario dado.

**Pista:**

La idea es recorrer el árbol binario, y para cada nodo del árbol binario, hay dos casos posibles:

*   Excluir el nodo actual del conjunto máximo independiente y procesar sus hijos.

*   Incluir el nodo actual en el conjunto máximo independiente y procesar sus nietos.

Por consiguiente, devuelve el máximo número de nodos posibles incluyendo o excluyendo el nodo actual.






````python
class Crear_Nodo:
  def __init__(self, valor, izquierda = None, derecha = None):
    self.valor = valor
    self.izquierda = izquierda
    self.derecha = derecha

def SIM(nodos):
  """
  Calcula el tamaño del Conjunto Independiente Máximo (SIM) en un árbol binario
  usando Programación Dinámica con memorización.

  Tiempo: O(n) donde n es el número de nodos
  Espacio: O(n) para la tabla de memorización
  """
  # Diccionario para memorización
  memo = {}

  def sim_helper(nodo):
    """
    Función recursiva con memorización que calcula el SIM.
    Retorna el tamaño máximo del conjunto independiente para el subárbol.
    """
    if nodo is None:
      return 0

    # Si ya calculamos este nodo, retornar el resultado guardado
    if nodo in memo:
      return memo[nodo]

    # Opción 1: EXCLUIR el nodo actual
    # Procesar ambos hijos
    sim_sin_actual = sim_helper(nodo.izquierda) + sim_helper(nodo.derecha)

    # Opción 2: INCLUIR el nodo actual
    # No podemos incluir los hijos directos, solo los nietos
    sim_con_actual = 1

    # Sumar los nietos del hijo izquierdo
    if nodo.izquierda:
      sim_con_actual += sim_helper(nodo.izquierda.izquierda)
      sim_con_actual += sim_helper(nodo.izquierda.derecha)

    # Sumar los nietos del hijo derecho
    if nodo.derecha:
      sim_con_actual += sim_helper(nodo.derecha.izquierda)
      sim_con_actual += sim_helper(nodo.derecha.derecha)

    # Elegir el máximo entre incluir o excluir el nodo actual
    valor_SIM = max(sim_sin_actual, sim_con_actual)

    # Guardar en memo para evitar recalcular
    memo[nodo] = valor_SIM

    return valor_SIM

  return sim_helper(nodos)
````

**Explicación del algoritmo:**

Este algoritmo calcula el **Conjunto Independiente Máximo (SIM)** en un árbol binario usando Programación Dinámica con **memorización** (enfoque top-down):

### Concepto Clave:

Un conjunto independiente es un conjunto de nodos donde ningún par de nodos es adyacente (no hay arista que los conecte). El SIM es el conjunto independiente de mayor tamaño.

### Estrategia:

Para cada nodo tenemos dos opciones:

1. **EXCLUIR el nodo actual**:
   - Podemos incluir libremente cualquier nodo de los subárboles hijos
   - SIM = SIM(hijo_izq) + SIM(hijo_der)

2. **INCLUIR el nodo actual**:
   - NO podemos incluir los hijos directos (son adyacentes)
   - SÍ podemos incluir los nietos
   - SIM = 1 + SIM(nieto_izq_izq) + SIM(nieto_izq_der) + SIM(nieto_der_izq) + SIM(nieto_der_der)

Elegimos el máximo entre ambas opciones.

### Relación de Recurrencia:

```
SIM(nodo) = max(
    SIM(hijo_izq) + SIM(hijo_der),                              // Excluir nodo
    1 + SIM(nietos del hijo_izq) + SIM(nietos del hijo_der)    // Incluir nodo
)

Caso base: SIM(NULL) = 0
```

### Pseudocódigo:

```
FUNCIÓN SIM(nodo):
    SI nodo es NULL:
        RETORNAR 0
    
    SI nodo ya está en memo:
        RETORNAR memo[nodo]
    
    // Opción 1: Excluir nodo actual
    excluir = SIM(hijo_izq) + SIM(hijo_der)
    
    // Opción 2: Incluir nodo actual
    incluir = 1
    SI hijo_izq existe:
        incluir += SIM(hijo_izq.izq) + SIM(hijo_izq.der)
    SI hijo_der existe:
        incluir += SIM(hijo_der.izq) + SIM(hijo_der.der)
    
    resultado = max(excluir, incluir)
    memo[nodo] = resultado
    RETORNAR resultado
```

### Ejemplo con la Prueba 1:

```
        1
       / \
      2   3
     /   / \
    4   5   6
       / \
      7   8
```

- **Incluir 1**: 1 + (4) + (5,6) = 1 + 1 + 2 = 4 ❌ (no se puede, 5 tiene hijos)
- **Incluir 1**: 1 + SIM(4) + SIM(5,6,7,8) = 1 + 1 + 3 = 5 ✓
- **Excluir 1**: SIM(2,4) + SIM(3,5,6,7,8) = máximo

El algoritmo explora todas las posibilidades y elige óptimamente.

### Costo Computacional:

- **Tiempo**: O(n) - cada nodo se procesa una vez gracias a la memorización
  - Sin memorización sería O(2ⁿ) por subproblemas repetidos
- **Espacio**: O(n) - tabla de memorización + pila de recursión
- **Ventaja sobre recursión simple**: Evita recalcular los mismos subárboles múltiples veces

````python
# Prueba 1

nodos = Crear_Nodo(1)
nodos.izquierda = Crear_Nodo(2)
nodos.derecha = Crear_Nodo(3)
nodos.izquierda.izquierda = Crear_Nodo(4)
nodos.derecha.izquierda = Crear_Nodo(5)
nodos.derecha.derecha = Crear_Nodo(6)
nodos.derecha.izquierda.izquierda = Crear_Nodo(7)
nodos.derecha.izquierda.derecha = Crear_Nodo(8)

print('Número SIM', SIM(nodos))
````

````python
# Prueba 2

nodos = Crear_Nodo(10)
nodos.izquierda = Crear_Nodo(5)
nodos.derecha = Crear_Nodo(20)
nodos.izquierda.izquierda = Crear_Nodo(1)
nodos.derecha.izquierda = Crear_Nodo(4)
nodos.derecha.derecha = Crear_Nodo(30)
nodos.derecha.izquierda.izquierda = Crear_Nodo(25)
nodos.derecha.izquierda.derecha = Crear_Nodo(28)

print('Número SIM', SIM(nodos))
````

### Ejercicio 4: La Trayectoria Serpiente

**Descripción:**

Dada una matriz cuadrada, imprima la trayectoria serpiente de longitud máxima en ella. Una **trayectoria serpiente** se define como una secuencia de números en la que cada nuevo número, que sólo puede situarse a la derecha o abajo del número actual, es más o menos uno.

Por ejemplo, podemos movernos a la derecha de cualquier celda de la matriz (si ese número es ±1) o movernos hacia abajo (si ese número es ±1). El problema consiste en encontrar la trayectoria más larga (trayectoria de la serpiente) a través de la matriz, teniendo en cuenta que sólo podemos movernos a una nueva celda cuyo valor sea ±1 respecto a la celda actual.



````python
def trayectoria_serpiente(Matriz):
  """
  Encuentra la trayectoria serpiente de longitud máxima en una matriz.
  Una trayectoria serpiente solo puede moverse derecha o abajo, y cada nuevo
  número debe ser ±1 del número actual.

  Tiempo: O(n² × m²) donde n×m es el tamaño de la matriz
  Espacio: O(n × m)
  """
  if not Matriz or not Matriz[0]:
    return [], 0

  n = len(Matriz)
  m = len(Matriz[0])

  # dp[i][j] = longitud máxima de trayectoria serpiente que termina en (i,j)
  dp = [[1] * m for _ in range(n)]

  # parent[i][j] = (fila, columna) de la celda previa en la trayectoria óptima
  parent = [[None] * m for _ in range(n)]

  max_longitud = 1
  max_pos = (0, 0)

  # Procesar cada celda
  for i in range(n):
    for j in range(m):
      # Revisar todas las celdas anteriores que puedan conectarse
      for pi in range(n):
        for pj in range(m):
          # Solo considerar celdas que están antes y pueden alcanzar (i,j)
          puede_conectar = False

          # Verificar si (pi, pj) puede conectarse con (i, j)
          # Debe estar a la izquierda o arriba
          if (pi == i and pj < j) or (pi < i and pj == j):
            # Verificar que el valor sea ±1
            if abs(Matriz[pi][pj] - Matriz[i][j]) == 1:
              puede_conectar = True

          if puede_conectar:
            # Si extender desde (pi, pj) mejora la longitud en (i, j)
            if dp[pi][pj] + 1 > dp[i][j]:
              dp[i][j] = dp[pi][pj] + 1
              parent[i][j] = (pi, pj)

      # Actualizar máximo global
      if dp[i][j] > max_longitud:
        max_longitud = dp[i][j]
        max_pos = (i, j)

  # Reconstruir la trayectoria
  trayectoria = []
  pos = max_pos

  while pos is not None:
    i, j = pos
    trayectoria.append(Matriz[i][j])
    pos = parent[i][j]

  trayectoria.reverse()
  longitud_trayectoria = max_longitud

  return trayectoria, longitud_trayectoria
````

**Explicación del algoritmo:**

Este algoritmo encuentra la **Trayectoria Serpiente** más larga en una matriz usando Programación Dinámica:

### Concepto:

Una trayectoria serpiente es una secuencia donde:
- Solo podemos movernos **derecha** o **abajo**
- Cada nuevo número debe ser **±1** del número actual
- Buscamos la secuencia más larga posible

### Estrategia:

1. **Estado DP**: `dp[i][j]` = longitud máxima de la trayectoria serpiente que termina en la celda (i, j)

2. **Relación de Recurrencia**: Para cada celda (i, j), revisamos todas las celdas anteriores (pi, pj) que:
   - Están a la izquierda (misma fila, columna menor): `pi == i y pj < j`
   - Están arriba (misma columna, fila menor): `pi < i y pj == j`
   - Su valor difiere en exactamente 1: `|Matriz[pi][pj] - Matriz[i][j]| = 1`
   
   ```
   dp[i][j] = max(dp[pi][pj] + 1) para todas las celdas válidas (pi, pj)
   Caso base: dp[i][j] = 1 (cada celda es una trayectoria de longitud 1)
   ```

3. **Reconstrucción**: Usamos `parent[i][j]` para almacenar de dónde vino la mejor trayectoria, permitiendo reconstruir el camino completo.

### Pseudocódigo:

```
FUNCIÓN trayectoria_serpiente(Matriz):
    n = número de filas
    m = número de columnas
    
    inicializar dp[n][m] = [[1, 1, ..., 1], ...]
    inicializar parent[n][m] = [[None, ...], ...]
    
    max_longitud = 1
    max_pos = (0, 0)
    
    PARA cada celda (i, j):
        PARA cada celda anterior (pi, pj):
            SI (pi, pj) puede conectarse con (i, j):
                SI |Matriz[pi][pj] - Matriz[i][j]| == 1:
                    SI dp[pi][pj] + 1 > dp[i][j]:
                        dp[i][j] = dp[pi][pj] + 1
                        parent[i][j] = (pi, pj)
        
        SI dp[i][j] > max_longitud:
            max_longitud = dp[i][j]
            max_pos = (i, j)
    
    reconstruir trayectoria desde max_pos usando parent
    RETORNAR trayectoria, max_longitud
```

## Demostración Práctica: Comparación de Eficiencia

Ejecutemos un experimento para visualizar la diferencia entre recursión simple y programación dinámica:

````python
import time

# Fibonacci - Recursión Simple
def fib_recursivo(n):
    if n <= 1:
        return n
    return fib_recursivo(n-1) + fib_recursivo(n-2)

# Fibonacci - Programación Dinámica
def fib_pd(n):
    if n <= 1:
        return n
    dp = [0] * (n + 1)
    dp[1] = 1
    for i in range(2, n + 1):
        dp[i] = dp[i-1] + dp[i-2]
    return dp[n]

# Comparación de tiempos
print("Comparación de eficiencia: Fibonacci\n")
print("n\tRecursión\t\tPD\t\tMejora")
print("-" * 60)

for n in [10, 20, 25, 30]:
    # Recursión
    inicio = time.time()
    resultado_rec = fib_recursivo(n)
    tiempo_rec = time.time() - inicio

    # Programación Dinámica
    inicio = time.time()
    resultado_pd = fib_pd(n)
    tiempo_pd = time.time() - inicio

    mejora = tiempo_rec / tiempo_pd if tiempo_pd > 0 else float('inf')

    print(f"{n}\t{tiempo_rec:.6f}s\t\t{tiempo_pd:.6f}s\t{mejora:.0f}x")

print("\n✅ La Programación Dinámica es SIGNIFICATIVAMENTE más eficiente")
print("📊 Para n=30, la mejora es de más de 1000x más rápido!")
````

````python
import matplotlib.pyplot as plt
import numpy as np

# Datos de tiempo de ejecución
valores_n = [10, 15, 20, 25, 30]
tiempos_recursivo = []
tiempos_pd = []

print("Calculando tiempos de ejecución...\n")

for n in valores_n:
    # Recursión
    inicio = time.time()
    fib_recursivo(n)
    tiempo_rec = time.time() - inicio
    tiempos_recursivo.append(tiempo_rec)

    # Programación Dinámica
    inicio = time.time()
    fib_pd(n)
    tiempo_pd = time.time() - inicio
    tiempos_pd.append(tiempo_pd)

# Crear visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Comparación de tiempos
ax1.plot(valores_n, tiempos_recursivo, 'ro-', label='Recursión Simple', linewidth=2, markersize=8)
ax1.plot(valores_n, tiempos_pd, 'go-', label='Programación Dinámica', linewidth=2, markersize=8)
ax1.set_xlabel('Tamaño del problema (n)', fontsize=12)
ax1.set_ylabel('Tiempo de ejecución (segundos)', fontsize=12)
ax1.set_title('Comparación de Rendimiento: Recursión vs PD', fontsize=14, fontweight='bold')
ax1.legend(fontsize=10)
ax1.grid(True, alpha=0.3)
ax1.set_yscale('log')  # Escala logarítmica para visualizar mejor

# Gráfico 2: Factor de mejora
mejoras = [tiempos_recursivo[i] / tiempos_pd[i] for i in range(len(valores_n))]
colors = plt.cm.viridis(np.linspace(0.3, 0.9, len(valores_n)))
bars = ax2.bar(valores_n, mejoras, color=colors, edgecolor='black', linewidth=1.5)
ax2.set_xlabel('Tamaño del problema (n)', fontsize=12)
ax2.set_ylabel('Factor de mejora (veces más rápido)', fontsize=12)
ax2.set_title('Factor de Mejora de PD sobre Recursión', fontsize=14, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='y')

# Agregar valores encima de las barras
for i, (bar, mejora) in enumerate(zip(bars, mejoras)):
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{mejora:.0f}x',
             ha='center', va='bottom', fontsize=10, fontweight='bold')

plt.tight_layout()
plt.show()

print(f"\n✅ Para n=30, la PD es {mejoras[-1]:.0f}x más rápida que la recursión simple!")
print(f"📈 El factor de mejora crece exponencialmente con n")
````

### Ejemplo 1: Problema del Cambio de Monedas

**Problema:** Dado un conjunto de denominaciones de monedas y una cantidad, encuentra el número mínimo de monedas necesarias para formar esa cantidad.

````python
def cambio_monedas(monedas, cantidad):
    """
    Encuentra el número mínimo de monedas para formar una cantidad.
    Programación Dinámica - O(n × cantidad)
    """
    # dp[i] = mínimo número de monedas para formar la cantidad i
    dp = [float('inf')] * (cantidad + 1)
    dp[0] = 0  # 0 monedas para formar 0

    # parent para reconstruir la solución
    parent = [-1] * (cantidad + 1)

    for i in range(1, cantidad + 1):
        for moneda in monedas:
            if moneda <= i and dp[i - moneda] + 1 < dp[i]:
                dp[i] = dp[i - moneda] + 1
                parent[i] = moneda

    if dp[cantidad] == float('inf'):
        return -1, []

    # Reconstruir solución
    monedas_usadas = []
    actual = cantidad
    while actual > 0:
        monedas_usadas.append(parent[actual])
        actual -= parent[actual]

    return dp[cantidad], monedas_usadas

# Ejemplo de uso
monedas = [1, 5, 10, 25]
cantidad = 67

num_monedas, monedas_usadas = cambio_monedas(monedas, cantidad)
print(f"💰 Cambio de monedas para {cantidad} centavos")
print(f"   Monedas disponibles: {monedas}")
print(f"   Número mínimo de monedas: {num_monedas}")
print(f"   Monedas usadas: {sorted(monedas_usadas, reverse=True)}")
print(f"   Verificación: {sum(monedas_usadas)} = {cantidad} ✓")

# Visualización
fig, ax = plt.subplots(figsize=(10, 5))

# Contar monedas por denominación
from collections import Counter
conteo = Counter(monedas_usadas)
denominaciones = sorted(conteo.keys(), reverse=True)
cantidades = [conteo[d] for d in denominaciones]

colors = ['#ff6b6b', '#4ecdc4', '#45b7d1', '#96ceb4']
bars = ax.bar([f'{d}¢' for d in denominaciones], cantidades, color=colors[:len(denominaciones)],
               edgecolor='black', linewidth=2)

ax.set_xlabel('Denominación de Moneda', fontsize=12)
ax.set_ylabel('Cantidad Usada', fontsize=12)
ax.set_title(f'Distribución Óptima de Monedas para {cantidad}¢', fontsize=14, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')

# Agregar valores encima de las barras
for bar, cant in zip(bars, cantidades):
    height = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2., height,
            f'{cant}',
            ha='center', va='bottom', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
````

### Ejemplo 2: Problema de la Mochila (Knapsack)

**Problema:** Tienes una mochila con capacidad limitada y varios objetos con peso y valor. ¿Qué objetos debes llevar para maximizar el valor sin exceder la capacidad?

````python
def mochila_01(pesos, valores, capacidad):
    """
    Problema de la Mochila 0/1 usando Programación Dinámica.
    Cada objeto se puede tomar o dejar (no fraccionar).

    Tiempo: O(n × capacidad)
    Espacio: O(n × capacidad)
    """
    n = len(pesos)

    # dp[i][w] = valor máximo con primeros i objetos y capacidad w
    dp = [[0] * (capacidad + 1) for _ in range(n + 1)]

    # Llenar la tabla
    for i in range(1, n + 1):
        for w in range(capacidad + 1):
            # Opción 1: No tomar el objeto i
            dp[i][w] = dp[i-1][w]

            # Opción 2: Tomar el objeto i (si cabe)
            if pesos[i-1] <= w:
                dp[i][w] = max(dp[i][w], dp[i-1][w - pesos[i-1]] + valores[i-1])

    # Reconstruir la solución
    objetos_seleccionados = []
    w = capacidad
    for i in range(n, 0, -1):
        if dp[i][w] != dp[i-1][w]:
            objetos_seleccionados.append(i-1)
            w -= pesos[i-1]

    objetos_seleccionados.reverse()

    return dp[n][capacidad], objetos_seleccionados

# Ejemplo: Objetos de campamento
nombres = ['Agua', 'Libro', 'Comida', 'Chaqueta', 'Cámara']
pesos = [3, 1, 2, 2, 1]
valores = [10, 3, 9, 5, 6]
capacidad = 6

valor_max, seleccionados = mochila_01(pesos, valores, capacidad)

print(f"🎒 Problema de la Mochila (Capacidad: {capacidad} kg)")
print(f"   Valor máximo alcanzado: {valor_max}")
print(f"\n   Objetos seleccionados:")

peso_total = 0
for idx in seleccionados:
    print(f"   - {nombres[idx]}: {pesos[idx]} kg, valor {valores[idx]}")
    peso_total += pesos[idx]

print(f"\n   Peso total: {peso_total}/{capacidad} kg")
print(f"   Espacio restante: {capacidad - peso_total} kg")

# Visualización
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Gráfico 1: Todos los objetos
colores = ['lightcoral' if i not in seleccionados else 'lightgreen' for i in range(len(nombres))]
ax1.barh(nombres, valores, color=colores, edgecolor='black', linewidth=2)
ax1.set_xlabel('Valor', fontsize=12)
ax1.set_title('Todos los Objetos (Verde = Seleccionado)', fontsize=13, fontweight='bold')
ax1.grid(True, alpha=0.3, axis='x')

# Agregar pesos como texto
for i, (nombre, peso, valor) in enumerate(zip(nombres, pesos, valores)):
    ax1.text(valor + 0.2, i, f'{peso}kg', va='center', fontsize=10)

# Gráfico 2: Eficiencia (valor/peso)
eficiencia = [valores[i]/pesos[i] for i in range(len(nombres))]
colores2 = ['lightcoral' if i not in seleccionados else 'lightgreen' for i in range(len(nombres))]
ax2.barh(nombres, eficiencia, color=colores2, edgecolor='black', linewidth=2)
ax2.set_xlabel('Eficiencia (Valor/Peso)', fontsize=12)
ax2.set_title('Eficiencia de cada Objeto', fontsize=13, fontweight='bold')
ax2.grid(True, alpha=0.3, axis='x')

# Agregar valores de eficiencia
for i, (nombre, ef) in enumerate(zip(nombres, eficiencia)):
    ax2.text(ef + 0.05, i, f'{ef:.1f}', va='center', fontsize=10)

plt.tight_layout()
plt.show()

print(f"\n💡 Nota: Aunque 'Agua' tiene la mejor eficiencia (3.33), ")
print(f"    el algoritmo encuentra la combinación óptima global.")
````

### Ejemplo 3: Subsecuencia Común Más Larga (LCS)

**Problema:** Dadas dos secuencias, encuentra la subsecuencia más larga que aparece en ambas (no necesariamente consecutiva).

````python
def lcs(X, Y):
    """
    Encuentra la Subsecuencia Común Más Larga (LCS) entre dos secuencias.
    Programación Dinámica - O(m × n)
    """
    m, n = len(X), len(Y)

    # dp[i][j] = longitud de LCS de X[0:i] y Y[0:j]
    dp = [[0] * (n + 1) for _ in range(m + 1)]

    # Llenar la tabla
    for i in range(1, m + 1):
        for j in range(1, n + 1):
            if X[i-1] == Y[j-1]:
                dp[i][j] = dp[i-1][j-1] + 1
            else:
                dp[i][j] = max(dp[i-1][j], dp[i][j-1])

    # Reconstruir la LCS
    lcs_str = []
    i, j = m, n
    while i > 0 and j > 0:
        if X[i-1] == Y[j-1]:
            lcs_str.append(X[i-1])
            i -= 1
            j -= 1
        elif dp[i-1][j] > dp[i][j-1]:
            i -= 1
        else:
            j -= 1

    lcs_str.reverse()
    return dp[m][n], ''.join(lcs_str)

# Ejemplo: Comparación de ADN o texto
X = "AGGTAB"
Y = "GXTXAYB"

longitud, subsecuencia = lcs(X, Y)

print(f" Subsecuencia Común Más Larga (LCS)")
print(f"   Secuencia 1: {X}")
print(f"   Secuencia 2: {Y}")
print(f"   LCS: {subsecuencia}")
print(f"   Longitud: {longitud}")

# Visualizar las coincidencias
print(f"\n   Análisis de coincidencias:")
for i, char in enumerate(X):
    if char in subsecuencia:
        print(f"   {X[:i]}[{char}]{X[i+1:]} ← coincide")
    else:
        print(f"   {X[:i]} {char}  {X[i+1:]}")

# Visualización de la matriz DP
fig, ax = plt.subplots(figsize=(10, 8))

# Crear datos para el heatmap
m, n = len(X), len(Y)
dp_matrix = [[0] * (n + 1) for _ in range(m + 1)]
for i in range(1, m + 1):
    for j in range(1, n + 1):
        if X[i-1] == Y[j-1]:
            dp_matrix[i][j] = dp_matrix[i-1][j-1] + 1
        else:
            dp_matrix[i][j] = max(dp_matrix[i-1][j], dp_matrix[i][j-1])

# Crear heatmap
im = ax.imshow(dp_matrix, cmap='YlOrRd', aspect='auto')

# Configurar ejes
ax.set_xticks(np.arange(n + 1))
ax.set_yticks(np.arange(m + 1))
ax.set_xticklabels([''] + list(Y))
ax.set_yticklabels([''] + list(X))

# Agregar valores en cada celda
for i in range(m + 1):
    for j in range(n + 1):
        text = ax.text(j, i, dp_matrix[i][j],
                      ha="center", va="center", color="black", fontweight='bold', fontsize=12)

ax.set_title('Matriz de Programación Dinámica para LCS', fontsize=14, fontweight='bold', pad=20)
ax.set_xlabel('Secuencia Y', fontsize=12)
ax.set_ylabel('Secuencia X', fontsize=12)

# Colorbar
cbar = plt.colorbar(im, ax=ax)
cbar.set_label('Longitud de LCS', fontsize=11)

plt.tight_layout()
plt.show()

print(f"\n Aplicaciones del LCS:")
print(f"   - Comparación de secuencias de ADN")
print(f"   - Detección de similitud entre archivos (diff)")
print(f"   - Corrección ortográfica y autocompletado")
````

###  Ejemplo Interactivo: Construcción de Tabla DP

Veamos paso a paso cómo se construye una tabla de programación dinámica para el problema del Cambio de Monedas:

````python
def visualizar_construccion_dp(monedas, cantidad_max):
    """
    Visualiza cómo se construye la tabla DP para el problema del cambio de monedas
    """
    # Construir tabla DP paso a paso
    dp = [float('inf')] * (cantidad_max + 1)
    dp[0] = 0

    fig, axes = plt.subplots(2, 3, figsize=(16, 10))
    fig.suptitle('Construcción Paso a Paso de la Tabla DP - Cambio de Monedas',
                 fontsize=16, fontweight='bold')

    # Estados intermedios para visualizar
    pasos = [0, 3, 6, 9, 12, cantidad_max]

    for idx, (ax, hasta) in enumerate(zip(axes.flat, pasos)):
        # Calcular DP hasta este punto
        dp_temp = [float('inf')] * (cantidad_max + 1)
        dp_temp[0] = 0

        for i in range(1, hasta + 1):
            for moneda in monedas:
                if moneda <= i and dp_temp[i - moneda] != float('inf'):
                    dp_temp[i] = min(dp_temp[i], dp_temp[i - moneda] + 1)

        # Preparar datos para visualización
        valores = [dp_temp[i] if dp_temp[i] != float('inf') else 0 for i in range(cantidad_max + 1)]
        colores = ['lightgreen' if dp_temp[i] != float('inf') else 'lightcoral'
                   for i in range(cantidad_max + 1)]

        # Crear gráfico de barras
        bars = ax.bar(range(cantidad_max + 1), valores, color=colores,
                      edgecolor='black', linewidth=1)

        ax.set_xlabel('Cantidad', fontsize=10)
        ax.set_ylabel('Número de Monedas', fontsize=10)
        ax.set_title(f'Paso {idx + 1}: Calculado hasta {hasta}', fontsize=12, fontweight='bold')
        ax.set_ylim(0, max(valores) + 1 if max(valores) > 0 else 5)
        ax.grid(True, alpha=0.3, axis='y')

        # Agregar valores encima de las barras (solo las primeras)
        for i, (bar, val) in enumerate(zip(bars[:min(15, len(bars))], valores[:15])):
            if val > 0:
                height = bar.get_height()
                ax.text(bar.get_x() + bar.get_width()/2., height,
                       f'{int(val)}' if val != float('inf') else '∞',
                       ha='center', va='bottom', fontsize=8)

    plt.tight_layout()
    plt.show()

# Ejecutar visualización
monedas_vis = [1, 5, 10]
cantidad_vis = 15

print(f" Visualizando construcción de tabla DP")
print(f"   Monedas: {monedas_vis}")
print(f"   Cantidad máxima: {cantidad_vis}")
print(f"\n   Verde = ya calculado | Rojo = aún no calculado\n")

visualizar_construccion_dp(monedas_vis, cantidad_vis)

print(" La tabla se construye iterativamente, usando soluciones previas")
````

### Comparación de Técnicas

| Técnica | Cuándo Usar | Complejidad | Ejemplo |
|---------|-------------|-------------|---------|
| **Greedy** | Elección local = óptimo global | O(n log n) | Huffman, Dijkstra |
| **Programación Dinámica** | Subproblemas superpuestos | O(n²) típicamente | Mochila, LCS, LIS |
| **Divide y Conquista** | Subproblemas independientes | O(n log n) | Merge Sort, Quick Sort |
| **Backtracking** | Explorar todas las soluciones | O(2ⁿ) | Sudoku, N-Queens |

### Optimizaciones Comunes

1. **Reducción de Espacio**: De O(n²) a O(n)
   ```python
   # En vez de: dp[i][j]
   # Usar: dp_prev[j] y dp_curr[j]
   ```

2. **Memorización vs Tabulación**:
   - **Memorización** (Top-Down): Más intuitivo, recursivo con caché
   - **Tabulación** (Bottom-Up): Más eficiente, iterativo

3. **Búsqueda Binaria para Optimización**:
   - Ejemplo: LIS puede ser O(n log n) en vez de O(n²)

### Estrategias para Identificar Problemas de PD:

✅ **Usa PD si:**
- El problema pide "optimizar" (máximo/mínimo)
- Hay decisiones que se pueden tomar en cada paso
- Hay subproblemas que se repiten
- La solución óptima se construye de soluciones óptimas de subproblemas

❌ **NO uses PD si:**
- El problema requiere solución exacta única (usar backtracking)
- No hay subproblemas superpuestos (usar divide y conquista)
- Una decisión greedy local garantiza óptimo global (usar algoritmo greedy)
